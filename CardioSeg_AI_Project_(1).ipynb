```python
import cv2
import numpy as np

def preprocess_slice(image_slice, mask_slice, target_size=(128, 128)):

    # ---------- MRI preprocessing ----------
    image_slice = image_slice.astype(np.float32)

    min_value = np.min(image_slice)
    max_value = np.max(image_slice)

    if max_value > min_value:
        image_normalized = (
            image_slice - min_value
        ) / (max_value - min_value)
    else:
        image_normalized = np.zeros_like(image_slice)

    # Resize MRI using linear interpolation
    image_resized = cv2.resize(
        image_normalized,
        target_size,
        interpolation=cv2.INTER_LINEAR
    )

    # Add channel dimension
    image_final = np.expand_dims(image_resized, axis=-1)

    # ---------- Mask preprocessing ----------
    mask_slice = mask_slice.astype(np.uint8)

    # Resize mask using nearest-neighbor
    mask_resized = cv2.resize(
        mask_slice,
        target_size,
        interpolation=cv2.INTER_NEAREST
    )

    mask_final = mask_resized.astype(np.uint8)

    return image_final, mask_final
```

```python
from tensorflow.keras import layers, Model

INPUT_SHAPE = (128, 128, 1)
NUM_CLASSES = 4

inputs = layers.Input(shape=INPUT_SHAPE)

print("Input shape:", inputs.shape)
print("Number of classes:", NUM_CLASSES)
```

```python
print("Input:", INPUT_SHAPE)
print("Output:", (128, 128, NUM_CLASSES))
print("Classes:", NUM_CLASSES)
```

```python
# Encoder Block 1

conv1 = layers.Conv2D(32, 3, padding="same", activation="relu")(inputs)
conv1 = layers.Conv2D(32, 3, padding="same", activation="relu")(conv1)

pool1 = layers.MaxPooling2D(pool_size=(2, 2))(conv1)

print("Block 1 convolution shape:", conv1.shape)
print("Block 1 pooled shape:", pool1.shape)
```

```python
# Encoder Block 2

conv2 = layers.Conv2D(64, 3, padding="same", activation="relu")(pool1)
conv2 = layers.Conv2D(64, 3, padding="same", activation="relu")(conv2)

pool2 = layers.MaxPooling2D(pool_size=(2, 2))(conv2)

print("Block 2 convolution shape:", conv2.shape)
print("Block 2 pooled shape:", pool2.shape)
```

```python
# Encoder Block 3

conv3 = layers.Conv2D(128, 3, padding="same", activation="relu")(pool2)
conv3 = layers.Conv2D(128, 3, padding="same", activation="relu")(conv3)

pool3 = layers.MaxPooling2D(pool_size=(2, 2))(conv3)

print("Block 3 convolution shape:", conv3.shape)
print("Block 3 pooled shape:", pool3.shape)
```

```python
# Encoder Block 4

conv4 = layers.Conv2D(256, 3, padding="same", activation="relu")(pool3)
conv4 = layers.Conv2D(256, 3, padding="same", activation="relu")(conv4)

pool4 = layers.MaxPooling2D(pool_size=(2, 2))(conv4)

print("Block 4 convolution shape:", conv4.shape)
print("Block 4 pooled shape:", pool4.shape)
```

```python
# Bottleneck

bottleneck = layers.Conv2D(
    512,
    3,
    padding="same",
    activation="relu"
)(pool4)

bottleneck = layers.Conv2D(
    512,
    3,
    padding="same",
    activation="relu"
)(bottleneck)

print("Bottleneck shape:", bottleneck.shape)
```

```python
# Decoder Block 1

up1 = layers.Conv2DTranspose(
    256,
    kernel_size=2,
    strides=2,
    padding="same"
)(bottleneck)

# Skip connection from Encoder Block 4
concat1 = layers.Concatenate()([up1, conv4])

dec1 = layers.Conv2D(
    256,
    3,
    padding="same",
    activation="relu"
)(concat1)

dec1 = layers.Conv2D(
    256,
    3,
    padding="same",
    activation="relu"
)(dec1)

print("Upsampled shape:", up1.shape)
print("After skip connection:", concat1.shape)
print("Decoder Block 1 shape:", dec1.shape)
```

```python
# Decoder Block 2

up2 = layers.Conv2DTranspose(
    128,
    kernel_size=2,
    strides=2,
    padding="same"
)(dec1)

# Skip connection from Encoder Block 3
concat2 = layers.Concatenate()([up2, conv3])

dec2 = layers.Conv2D(
    128,
    3,
    padding="same",
    activation="relu"
)(concat2)

dec2 = layers.Conv2D(
    128,
    3,
    padding="same",
    activation="relu"
)(dec2)

print("Upsampled shape:", up2.shape)
print("After skip connection:", concat2.shape)
print("Decoder Block 2 shape:", dec2.shape)
```

```python
# Decoder Block 3

up3 = layers.Conv2DTranspose(
    64,
    kernel_size=2,
    strides=2,
    padding="same"
)(dec2)

# Skip connection from Encoder Block 2
concat3 = layers.Concatenate()([up3, conv2])

dec3 = layers.Conv2D(
    64,
    3,
    padding="same",
    activation="relu"
)(concat3)

dec3 = layers.Conv2D(
    64,
    3,
    padding="same",
    activation="relu"
)(dec3)

print("Upsampled shape:", up3.shape)
print("After skip connection:", concat3.shape)
print("Decoder Block 3 shape:", dec3.shape)
```

```python
# Decoder Block 4

up4 = layers.Conv2DTranspose(
    32,
    kernel_size=2,
    strides=2,
    padding="same"
)(dec3)

# Skip connection from Encoder Block 1
concat4 = layers.Concatenate()([up4, conv1])

dec4 = layers.Conv2D(
    32,
    3,
    padding="same",
    activation="relu"
)(concat4)

dec4 = layers.Conv2D(
    32,
    3,
    padding="same",
    activation="relu"
)(dec4)

print("Upsampled shape:", up4.shape)
print("After skip connection:", concat4.shape)
print("Decoder Block 4 shape:", dec4.shape)
```

```python
# Final output layer

outputs = layers.Conv2D(
    NUM_CLASSES,
    kernel_size=1,
    activation="softmax",
    name="segmentation_output"
)(dec4)

print("Final output shape:", outputs.shape)
```

```python
model = Model(
    inputs=inputs,
    outputs=outputs,
    name="CardioSeg_AI_UNet"
)

print("Model created successfully!")
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)
```

```python
import tensorflow as tf

def dice_coefficient(y_true, y_pred, smooth=1e-6):
    """
    Multi-class Dice coefficient.
    """

    # Convert ground-truth labels to one-hot format
    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    # Flatten
    y_true = tf.reshape(y_true, [-1, NUM_CLASSES])
    y_pred = tf.reshape(y_pred, [-1, NUM_CLASSES])

    # Calculate intersection
    intersection = tf.reduce_sum(y_true * y_pred, axis=0)

    # Calculate Dice for each class
    denominator = (
        tf.reduce_sum(y_true, axis=0) +
        tf.reduce_sum(y_pred, axis=0)
    )

    dice = (2.0 * intersection + smooth) / (denominator + smooth)

    # Average Dice across 4 classes
    return tf.reduce_mean(dice)
```

```python
print("Dice function created successfully!")
```

```python
import tensorflow as tf

def weighted_dice_loss(y_true, y_pred, smooth=1e-6):
    """
    Weighted multi-class Dice loss.
    Gives more importance to cardiac structures
    than the background.
    """

    # Convert ground-truth mask to one-hot
    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    # Flatten
    y_true = tf.reshape(y_true, [-1, NUM_CLASSES])
    y_pred = tf.reshape(y_pred, [-1, NUM_CLASSES])

    # Calculate intersection
    intersection = tf.reduce_sum(y_true * y_pred, axis=0)

    # Calculate denominator
    denominator = (
        tf.reduce_sum(y_true, axis=0) +
        tf.reduce_sum(y_pred, axis=0)
    )

    # Dice for each class
    dice = (
        2.0 * intersection + smooth
    ) / (
        denominator + smooth
    )

    # Class weights
    weights = tf.constant(
        [0.2, 0.3, 0.3, 0.2],
        dtype=tf.float32
    )

    # Weighted Dice
    weighted_dice = tf.reduce_sum(dice * weights)

    return 1.0 - weighted_dice
```

```python
def categorical_cross_entropy_loss(y_true, y_pred):
    """
    Multi-class categorical cross-entropy.
    """

    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    loss = tf.keras.losses.categorical_crossentropy(
        y_true,
        y_pred
    )

    return tf.reduce_mean(loss)
```

```python
def hybrid_loss(y_true, y_pred):
    dice_loss = weighted_dice_loss(y_true, y_pred)
    ce_loss = categorical_cross_entropy_loss(y_true, y_pred)

    return 0.7 * dice_loss + 0.3 * ce_loss
```

```python
print("Hybrid loss function created successfully!")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# -----------------------------
# Model configuration
# -----------------------------
INPUT_SHAPE = (128, 128, 1)
NUM_CLASSES = 4

# -----------------------------
# Input
# -----------------------------
inputs = layers.Input(
    shape=INPUT_SHAPE,
    name="mri_input"
)

# -----------------------------
# Encoder Block 1
# -----------------------------
conv1 = layers.Conv2D(
    32, 3, padding="same", activation="relu"
)(inputs)
conv1 = layers.Conv2D(
    32, 3, padding="same", activation="relu"
)(conv1)
pool1 = layers.MaxPooling2D(2)(conv1)

# -----------------------------
# Encoder Block 2
# -----------------------------
conv2 = layers.Conv2D(
    64, 3, padding="same", activation="relu"
)(pool1)
conv2 = layers.Conv2D(
    64, 3, padding="same", activation="relu"
)(conv2)
pool2 = layers.MaxPooling2D(2)(conv2)

# -----------------------------
# Encoder Block 3
# -----------------------------
conv3 = layers.Conv2D(
    128, 3, padding="same", activation="relu"
)(pool2)
conv3 = layers.Conv2D(
    128, 3, padding="same", activation="relu"
)(conv3)
pool3 = layers.MaxPooling2D(2)(conv3)

# -----------------------------
# Encoder Block 4
# -----------------------------
conv4 = layers.Conv2D(
    256, 3, padding="same", activation="relu"
)(pool3)
conv4 = layers.Conv2D(
    256, 3, padding="same", activation="relu"
)(conv4)
pool4 = layers.MaxPooling2D(2)(conv4)

# -----------------------------
# Bottleneck
# -----------------------------
bottleneck = layers.Conv2D(
    512, 3, padding="same", activation="relu"
)(pool4)
bottleneck = layers.Conv2D(
    512, 3, padding="same", activation="relu"
)(bottleneck)

# -----------------------------
# Decoder Block 1
# -----------------------------
up1 = layers.Conv2DTranspose(
    256, 2, strides=2, padding="same"
)(bottleneck)

concat1 = layers.Concatenate()([
    up1,
    conv4
])

dec1 = layers.Conv2D(
    256, 3, padding="same", activation="relu"
)(concat1)
dec1 = layers.Conv2D(
    256, 3, padding="same", activation="relu"
)(dec1)

# -----------------------------
# Decoder Block 2
# -----------------------------
up2 = layers.Conv2DTranspose(
    128, 2, strides=2, padding="same"
)(dec1)

concat2 = layers.Concatenate()([
    up2,
    conv3
])

dec2 = layers.Conv2D(
    128, 3, padding="same", activation="relu"
)(concat2)
dec2 = layers.Conv2D(
    128, 3, padding="same", activation="relu"
)(dec2)

# -----------------------------
# Decoder Block 3
# -----------------------------
up3 = layers.Conv2DTranspose(
    64, 2, strides=2, padding="same"
)(dec2)

concat3 = layers.Concatenate()([
    up3,
    conv2
])

dec3 = layers.Conv2D(
    64, 3, padding="same", activation="relu"
)(concat3)
dec3 = layers.Conv2D(
    64, 3, padding="same", activation="relu"
)(dec3)

# -----------------------------
# Decoder Block 4
# -----------------------------
up4 = layers.Conv2DTranspose(
    32, 2, strides=2, padding="same"
)(dec3)

concat4 = layers.Concatenate()([
    up4,
    conv1
])

dec4 = layers.Conv2D(
    32, 3, padding="same", activation="relu"
)(concat4)
dec4 = layers.Conv2D(
    32, 3, padding="same", activation="relu"
)(dec4)

# -----------------------------
# Final segmentation output
# -----------------------------
outputs = layers.Conv2D(
    NUM_CLASSES,
    kernel_size=1,
    activation="softmax",
    dtype="float32",
    name="segmentation_output"
)(dec4)

# -----------------------------
# Complete model
# -----------------------------
model = Model(
    inputs=inputs,
    outputs=outputs,
    name="CardioSeg_AI_UNet"
)

print("Model rebuilt successfully!")
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)
print("Output dtype:", model.output.dtype)
```

```python
import tensorflow as tf

NUM_CLASSES = 4

# -----------------------------
# Dice coefficient
# -----------------------------
def dice_coefficient(y_true, y_pred, smooth=1e-6):

    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    y_true = tf.reshape(
        y_true,
        [-1, NUM_CLASSES]
    )

    y_pred = tf.reshape(
        y_pred,
        [-1, NUM_CLASSES]
    )

    intersection = tf.reduce_sum(
        y_true * y_pred,
        axis=0
    )

    denominator = (
        tf.reduce_sum(y_true, axis=0) +
        tf.reduce_sum(y_pred, axis=0)
    )

    dice = (
        2.0 * intersection + smooth
    ) / (
        denominator + smooth
    )

    return tf.reduce_mean(dice)


# -----------------------------
# Weighted Dice Loss
# -----------------------------
def weighted_dice_loss(y_true, y_pred, smooth=1e-6):

    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    y_true = tf.reshape(
        y_true,
        [-1, NUM_CLASSES]
    )

    y_pred = tf.reshape(
        y_pred,
        [-1, NUM_CLASSES]
    )

    intersection = tf.reduce_sum(
        y_true * y_pred,
        axis=0
    )

    denominator = (
        tf.reduce_sum(y_true, axis=0) +
        tf.reduce_sum(y_pred, axis=0)
    )

    dice = (
        2.0 * intersection + smooth
    ) / (
        denominator + smooth
    )

    # Background, RV, Myocardium, LV
    weights = tf.constant(
        [0.2, 0.3, 0.3, 0.2],
        dtype=tf.float32
    )

    weighted_dice = tf.reduce_sum(
        dice * weights
    )

    return 1.0 - weighted_dice


# -----------------------------
# Categorical Cross Entropy
# -----------------------------
def categorical_cross_entropy_loss(y_true, y_pred):

    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    loss = tf.keras.losses.categorical_crossentropy(
        y_true,
        y_pred
    )

    return tf.reduce_mean(loss)


# -----------------------------
# Hybrid Loss
# -----------------------------
def hybrid_loss(y_true, y_pred):

    dice_loss = weighted_dice_loss(
        y_true,
        y_pred
    )

    ce_loss = categorical_cross_entropy_loss(
        y_true,
        y_pred
    )

    return (
        0.7 * dice_loss +
        0.3 * ce_loss
    )


print("Dice coefficient restored!")
print("Weighted Dice loss restored!")
print("Cross-entropy loss restored!")
print("Hybrid loss restored!")
```

```python
optimizer = tf.keras.optimizers.AdamW(
    learning_rate=0.002,
    weight_decay=1e-4
)

model.compile(
    optimizer=optimizer,
    loss=hybrid_loss,
    metrics=[dice_coefficient]
)

print("Model compiled successfully!")
```

```python
import os
import numpy as np
import nibabel as nib
import cv2

def get_labeled_frames(patient_dir):
    """
    Find MRI frames that have corresponding ground-truth masks.
    """
    files = os.listdir(patient_dir)

    labeled_frames = []

    for file in files:
        if file.endswith("_gt.nii.gz"):
            # Example:
            # patient001_frame01_gt.nii.gz
            frame_name = file.replace("_gt.nii.gz", "")

            image_file = frame_name + ".nii.gz"

            if image_file in files:
                labeled_frames.append(
                    (image_file, file)
                )

    return sorted(labeled_frames)


def load_patient_slices(patient_id):
    """
    Load all labeled 2D slices from one patient.
    Returns:
        images: (N, 128, 128, 1)
        masks:  (N, 128, 128)
    """

    patient_dir = os.path.join(TRAIN_DIR, patient_id)

    labeled_frames = get_labeled_frames(patient_dir)

    images = []
    masks = []

    for image_file, mask_file in labeled_frames:

        image_path = os.path.join(patient_dir, image_file)
        mask_path = os.path.join(patient_dir, mask_file)

        image_nii = nib.load(image_path)
        mask_nii = nib.load(mask_path)

        image_data = image_nii.get_fdata()
        mask_data = mask_nii.get_fdata()

        # Convert 3D volume into 2D slices
        num_slices = image_data.shape[2]

        for slice_index in range(num_slices):

            image_slice = image_data[:, :, slice_index]
            mask_slice = mask_data[:, :, slice_index]

            image_final, mask_final = preprocess_slice(
                image_slice,
                mask_slice,
                target_size=(128, 128)
            )

            images.append(image_final)
            masks.append(mask_final)

    return np.array(images, dtype=np.float32), np.array(masks, dtype=np.uint8)


print("Slice pipeline created successfully!")
```

```python
import cv2
import numpy as np

def preprocess_slice(image_slice, mask_slice, target_size=(128, 128)):

    # -------------------------
    # MRI preprocessing
    # -------------------------
    image_slice = image_slice.astype(np.float32)

    min_value = np.min(image_slice)
    max_value = np.max(image_slice)

    if max_value > min_value:
        image_normalized = (
            image_slice - min_value
        ) / (max_value - min_value)
    else:
        image_normalized = np.zeros_like(image_slice)

    # Resize MRI
    image_resized = cv2.resize(
        image_normalized,
        target_size,
        interpolation=cv2.INTER_LINEAR
    )

    # Add channel dimension
    image_final = np.expand_dims(
        image_resized,
        axis=-1
    )

    # -------------------------
    # Mask preprocessing
    # -------------------------
    mask_slice = mask_slice.astype(np.uint8)

    # IMPORTANT:
    # Nearest-neighbor keeps labels 0,1,2,3
    mask_resized = cv2.resize(
        mask_slice,
        target_size,
        interpolation=cv2.INTER_NEAREST
    )

    mask_final = mask_resized.astype(np.uint8)

    return image_final, mask_final


print("preprocess_slice() restored successfully!")
```

```python
import os
import numpy as np
import nibabel as nib
import tensorflow as tf


class CardioSegDataGenerator(tf.keras.utils.Sequence):

    def __init__(
        self,
        patient_list,
        batch_size=8,
        shuffle=True
    ):
        self.patient_list = patient_list
        self.batch_size = batch_size
        self.shuffle = shuffle

        self.samples = []

        for patient_id in self.patient_list:

            patient_dir = os.path.join(
                TRAIN_DIR,
                patient_id
            )

            labeled_frames = get_labeled_frames(
                patient_dir
            )

            for image_file, mask_file in labeled_frames:

                image_path = os.path.join(
                    patient_dir,
                    image_file
                )

                mask_path = os.path.join(
                    patient_dir,
                    mask_file
                )

                image_nii = nib.load(image_path)
                num_slices = image_nii.shape[2]

                for slice_index in range(num_slices):

                    self.samples.append(
                        (
                            image_path,
                            mask_path,
                            slice_index
                        )
                    )

        self.indices = np.arange(len(self.samples))
        self.on_epoch_end()

        print(
            f"Patients: {len(self.patient_list)} | "
            f"Slices: {len(self.samples)}"
        )

    def __len__(self):
        return int(
            np.ceil(
                len(self.samples) / self.batch_size
            )
        )

    def __getitem__(self, index):

        batch_indices = self.indices[
            index * self.batch_size:
            (index + 1) * self.batch_size
        ]

        batch_images = []
        batch_masks = []

        for idx in batch_indices:

            image_path, mask_path, slice_index = self.samples[idx]

            image_nii = nib.load(image_path)
            mask_nii = nib.load(mask_path)

            image_data = image_nii.get_fdata()
            mask_data = mask_nii.get_fdata()

            image_slice = image_data[:, :, slice_index]
            mask_slice = mask_data[:, :, slice_index]

            image_final, mask_final = preprocess_slice(
                image_slice,
                mask_slice,
                target_size=(128, 128)
            )

            batch_images.append(image_final)
            batch_masks.append(mask_final)

        return (
            np.asarray(batch_images, dtype=np.float32),
            np.asarray(batch_masks, dtype=np.uint8)
        )

    def on_epoch_end(self):

        if self.shuffle:
            np.random.shuffle(self.indices)


print("CardioSegDataGenerator restored successfully!")
```

```python
import numpy as np

# Class-wise Dice calculation
class_names = {
    0: "Background",
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

class_dice_scores = {i: [] for i in range(4)}

for batch_idx in range(len(test_generator)):
    images, true_masks = test_generator[batch_idx]

    predictions = model.predict(images, verbose=0)
    predicted_masks = np.argmax(predictions, axis=-1)

    for class_id in range(4):
        for true_mask, pred_mask in zip(true_masks, predicted_masks):

            true_class = (true_mask == class_id)
            pred_class = (pred_mask == class_id)

            intersection = np.sum(true_class & pred_class)
            denominator = np.sum(true_class) + np.sum(pred_class)

            if denominator == 0:
                dice = 1.0
            else:
                dice = (2.0 * intersection) / denominator

            class_dice_scores[class_id].append(dice)

# Print average Dice for each class
print("\n===== CLASS-WISE TEST DICE =====")

for class_id, name in class_names.items():
    score = np.mean(class_dice_scores[class_id])
    print(f"{name}: {score:.4f} ({score*100:.2f}%)")
```

```python
# Step 19: RV + Myocardium focused Dice loss

def improved_weighted_dice_loss(y_true, y_pred, smooth=1e-6):

    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    y_true = tf.reshape(y_true, [-1, NUM_CLASSES])
    y_pred = tf.reshape(y_pred, [-1, NUM_CLASSES])

    intersection = tf.reduce_sum(y_true * y_pred, axis=0)

    denominator = (
        tf.reduce_sum(y_true, axis=0)
        + tf.reduce_sum(y_pred, axis=0)
    )

    dice = (
        2.0 * intersection + smooth
    ) / (
        denominator + smooth
    )

    # Higher importance for RV and Myocardium
    weights = tf.constant(
        [0.10, 0.40, 0.35, 0.15],
        dtype=tf.float32
    )

    weighted_dice = tf.reduce_sum(dice * weights)

    return 1.0 - weighted_dice


def improved_hybrid_loss(y_true, y_pred):

    dice_loss = improved_weighted_dice_loss(y_true, y_pred)

    y_true_one_hot = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    ce_loss = tf.reduce_mean(
        tf.keras.losses.categorical_crossentropy(
            y_true_one_hot,
            y_pred
        )
    )

    return 0.7 * dice_loss + 0.3 * ce_loss


print("Improved RV + Myocardium focused loss created successfully.")
```

```python
# Step 20: Fine-tune model with RV + Myocardium focused loss

import tensorflow as tf
import os

# Load the original best model weights from Epoch 17
model.load_weights(checkpoint_path)

# Recompile with the improved loss
optimizer_improved = tf.keras.optimizers.AdamW(
    learning_rate=0.0005,
    weight_decay=1e-4
)

model.compile(
    optimizer=optimizer_improved,
    loss=improved_hybrid_loss,
    metrics=[dice_coefficient]
)

# Save improved model separately
improved_checkpoint_path = os.path.join(
    MODEL_DIR,
    "CardioSeg_AI_RV_Myo_best.keras"
)

improved_checkpoint = tf.keras.callbacks.ModelCheckpoint(
    improved_checkpoint_path,
    monitor="val_dice_coefficient",
    mode="max",
    save_best_only=True,
    verbose=1
)

improved_reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_dice_coefficient",
    mode="max",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

improved_early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_dice_coefficient",
    mode="max",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

# Fine-tune for a maximum of 15 epochs
improved_history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=15,
    callbacks=[
        improved_checkpoint,
        improved_reduce_lr,
        improved_early_stop
    ],
    verbose=1
)
```

```python
# Step 25B: Create controlled augmentation function

import cv2
import numpy as np

def augment_slice(image, mask):
    """
    Apply the same spatial transformation to MRI image and segmentation mask.
    """

    # Remove channel temporarily
    image_2d = image[:, :, 0].astype(np.float32)
    mask_2d = mask.astype(np.uint8)

    h, w = image_2d.shape

    # -----------------------------
    # 1. Small random rotation
    # -----------------------------
    angle = np.random.uniform(-10, 10)

    center = (w / 2, h / 2)

    rotation_matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    image_2d = cv2.warpAffine(
        image_2d,
        rotation_matrix,
        (w, h),
        flags=cv2.INTER_LINEAR,
        borderMode=cv2.BORDER_REFLECT
    )

    mask_2d = cv2.warpAffine(
        mask_2d,
        rotation_matrix,
        (w, h),
        flags=cv2.INTER_NEAREST,
        borderMode=cv2.BORDER_CONSTANT,
        borderValue=0
    )

    # -----------------------------
    # 2. Small translation
    # -----------------------------
    shift_x = np.random.uniform(-5, 5)
    shift_y = np.random.uniform(-5, 5)

    translation_matrix = np.float32([
        [1, 0, shift_x],
        [0, 1, shift_y]
    ])

    image_2d = cv2.warpAffine(
        image_2d,
        translation_matrix,
        (w, h),
        flags=cv2.INTER_LINEAR,
        borderMode=cv2.BORDER_REFLECT
    )

    mask_2d = cv2.warpAffine(
        mask_2d,
        translation_matrix,
        (w, h),
        flags=cv2.INTER_NEAREST,
        borderMode=cv2.BORDER_CONSTANT,
        borderValue=0
    )

    # -----------------------------
    # 3. Small zoom
    # -----------------------------
    zoom = np.random.uniform(0.95, 1.05)

    new_w = int(w * zoom)
    new_h = int(h * zoom)

    image_zoom = cv2.resize(
        image_2d,
        (new_w, new_h),
        interpolation=cv2.INTER_LINEAR
    )

    mask_zoom = cv2.resize(
        mask_2d,
        (new_w, new_h),
        interpolation=cv2.INTER_NEAREST
    )

    # Crop/pad back to original size
    def center_crop_or_pad(arr, target_h, target_w, is_mask=False):

        current_h, current_w = arr.shape

        # Crop
        if current_h > target_h:
            start_y = (current_h - target_h) // 2
            arr = arr[start_y:start_y + target_h, :]

        if current_w > target_w:
            start_x = (current_w - target_w) // 2
            arr = arr[:, start_x:start_x + target_w]

        # Pad
        current_h, current_w = arr.shape

        pad_y = max(0, target_h - current_h)
        pad_x = max(0, target_w - current_w)

        if pad_y > 0 or pad_x > 0:
            arr = np.pad(
                arr,
                (
                    (pad_y // 2, pad_y - pad_y // 2),
                    (pad_x // 2, pad_x - pad_x // 2)
                ),
                mode="constant",
                constant_values=0
            )

        return arr

    image_2d = center_crop_or_pad(
        image_zoom, h, w
    )

    mask_2d = center_crop_or_pad(
        mask_zoom, h, w
    )

    # -----------------------------
    # 4. Mild intensity variation
    # -----------------------------
    brightness = np.random.uniform(-0.05, 0.05)
    contrast = np.random.uniform(0.95, 1.05)

    image_2d = image_2d * contrast + brightness

    image_2d = np.clip(
        image_2d,
        0.0,
        1.0
    )

    # Restore channel dimension
    image_final = np.expand_dims(
        image_2d.astype(np.float32),
        axis=-1
    )

    mask_final = mask_2d.astype(np.uint8)

    return image_final, mask_final


print("Augmentation function created successfully.")
```

```python
# Step 25E: Augmented training data generator

class AugmentedCardioSegDataGenerator(tf.keras.utils.Sequence):

    def __init__(self, patient_list, batch_size=8, shuffle=True):
        self.patient_list = patient_list
        self.batch_size = batch_size
        self.shuffle = shuffle
        self.samples = []

        for patient_id in self.patient_list:

            patient_dir = os.path.join(TRAIN_DIR, patient_id)

            labeled_frames = get_labeled_frames(patient_dir)

            for image_file, mask_file in labeled_frames:

                image_path = os.path.join(
                    patient_dir,
                    image_file
                )

                mask_path = os.path.join(
                    patient_dir,
                    mask_file
                )

                image_nii = nib.load(image_path)

                num_slices = image_nii.shape[2]

                for slice_index in range(num_slices):

                    self.samples.append(
                        (
                            image_path,
                            mask_path,
                            slice_index
                        )
                    )

        self.indices = np.arange(len(self.samples))

        self.on_epoch_end()

        print(
            f"Patients: {len(self.patient_list)} | "
            f"Slices: {len(self.samples)}"
        )

    def __len__(self):

        return int(
            np.ceil(
                len(self.samples) /
                self.batch_size
            )
        )

    def __getitem__(self, index):

        batch_indices = self.indices[
            index * self.batch_size:
            (index + 1) * self.batch_size
        ]

        batch_images = []
        batch_masks = []

        for idx in batch_indices:

            image_path, mask_path, slice_index = \
                self.samples[idx]

            image_nii = nib.load(image_path)
            mask_nii = nib.load(mask_path)

            image_data = image_nii.get_fdata()
            mask_data = mask_nii.get_fdata()

            image_slice = image_data[:, :, slice_index]
            mask_slice = mask_data[:, :, slice_index]

            # Normal preprocessing
            image_final, mask_final = preprocess_slice(
                image_slice,
                mask_slice,
                target_size=(128, 128)
            )

            # Apply augmentation ONLY to training data
            image_final, mask_final = augment_slice(
                image_final,
                mask_final
            )

            batch_images.append(image_final)
            batch_masks.append(mask_final)

        return (
            np.asarray(batch_images, dtype=np.float32),
            np.asarray(batch_masks, dtype=np.uint8)
        )

    def on_epoch_end(self):

        if self.shuffle:
            np.random.shuffle(self.indices)


print("Augmented training generator created successfully.")
```

```python
# Step 26A: Boundary-aware loss

import tensorflow as tf

def boundary_aware_loss(y_true, y_pred, smooth=1e-6):

    # Convert ground-truth mask to one-hot
    y_true_one_hot = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    # -----------------------------
    # 1. Weighted Dice Loss
    # -----------------------------

    y_true_flat = tf.reshape(
        y_true_one_hot,
        [-1, NUM_CLASSES]
    )

    y_pred_flat = tf.reshape(
        y_pred,
        [-1, NUM_CLASSES]
    )

    intersection = tf.reduce_sum(
        y_true_flat * y_pred_flat,
        axis=0
    )

    denominator = (
        tf.reduce_sum(y_true_flat, axis=0)
        +
        tf.reduce_sum(y_pred_flat, axis=0)
    )

    dice = (
        2.0 * intersection + smooth
    ) / (
        denominator + smooth
    )

    # Give more importance to RV and myocardium
    weights = tf.constant(
        [0.10, 0.40, 0.35, 0.15],
        dtype=tf.float32
    )

    weighted_dice = tf.reduce_sum(
        dice * weights
    )

    dice_loss = 1.0 - weighted_dice

    # -----------------------------
    # 2. Cross Entropy
    # -----------------------------

    ce_loss = tf.keras.losses.categorical_crossentropy(
        y_true_one_hot,
        y_pred
    )

    ce_loss = tf.reduce_mean(ce_loss)

    # -----------------------------
    # 3. Boundary approximation
    # -----------------------------

    # Detect changes between neighboring pixels
    true_horizontal = tf.abs(
        y_true_one_hot[:, 1:, :, :]
        -
        y_true_one_hot[:, :-1, :, :]
    )

    true_vertical = tf.abs(
        y_true_one_hot[:, :, 1:, :]
        -
        y_true_one_hot[:, :, :-1, :]
    )

    pred_horizontal = tf.abs(
        y_pred[:, 1:, :, :]
        -
        y_pred[:, :-1, :, :]
    )

    pred_vertical = tf.abs(
        y_pred[:, :, 1:, :]
        -
        y_pred[:, :, :-1, :]
    )

    boundary_loss = (
        tf.reduce_mean(
            tf.abs(
                true_horizontal -
                pred_horizontal
            )
        )
        +
        tf.reduce_mean(
            tf.abs(
                true_vertical -
                pred_vertical
            )
        )
    )

    # -----------------------------
    # 4. Final combined loss
    # -----------------------------

    total_loss = (
        0.60 * dice_loss
        +
        0.30 * ce_loss
        +
        0.10 * boundary_loss
    )

    return total_loss


print("Boundary-aware loss created successfully.")
```

```python
# Step 26B: Compile model with boundary-aware loss

optimizer_boundary = tf.keras.optimizers.AdamW(
    learning_rate=0.0001,
    weight_decay=1e-4
)

model.compile(
    optimizer=optimizer_boundary,
    loss=boundary_aware_loss,
    metrics=[dice_coefficient]
)

boundary_checkpoint_path = os.path.join(
    MODEL_DIR,
    "CardioSeg_AI_Boundary_best.keras"
)

boundary_checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
    filepath=boundary_checkpoint_path,
    monitor="val_dice_coefficient",
    mode="max",
    save_best_only=True,
    verbose=1
)

boundary_reduce_lr_callback = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_dice_coefficient",
    mode="max",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

boundary_early_stopping_callback = tf.keras.callbacks.EarlyStopping(
    monitor="val_dice_coefficient",
    mode="max",
    patience=4,
    restore_best_weights=True,
    verbose=1
)

print("Boundary-aware model compiled successfully.")
print("Learning rate: 0.0001")
```

```python
# Step 27: Class-wise Dice for boundary-aware model

class_names = {
    0: "Background",
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

val_class_dice_boundary = {i: [] for i in range(4)}

for batch_idx in range(len(val_generator)):

    images, true_masks = val_generator[batch_idx]

    predictions = model.predict(images, verbose=0)
    predicted_masks = np.argmax(predictions, axis=-1)

    for class_id in range(4):

        for true_mask, pred_mask in zip(true_masks, predicted_masks):

            true_class = (true_mask == class_id)
            pred_class = (pred_mask == class_id)

            intersection = np.sum(
                true_class & pred_class
            )

            denominator = (
                np.sum(true_class)
                +
                np.sum(pred_class)
            )

            if denominator == 0:
                dice = 1.0
            else:
                dice = (
                    2.0 * intersection
                ) / denominator

            val_class_dice_boundary[class_id].append(dice)


print("\n===== BOUNDARY-AWARE VALIDATION DICE =====")

for class_id, name in class_names.items():

    score = np.mean(
        val_class_dice_boundary[class_id]
    )

    print(
        f"{name}: "
        f"{score:.4f} "
        f"({score*100:.2f}%)"
    )
```

```python
# Step 28: Analyze class distribution in the training dataset

import numpy as np
import nibabel as nib
import os

class_pixel_counts = {
    0: 0,  # Background
    1: 0,  # RV
    2: 0,  # Myocardium
    3: 0   # LV
}

slice_counts = {
    "total": 0,
    "background_only": 0,
    "RV": 0,
    "Myocardium": 0,
    "LV": 0,
    "all_foreground": 0
}

for patient_id in train_patients:

    patient_dir = os.path.join(
        TRAIN_DIR,
        patient_id
    )

    labeled_frames = get_labeled_frames(patient_dir)

    for image_file, mask_file in labeled_frames:

        mask_path = os.path.join(
            patient_dir,
            mask_file
        )

        mask_nii = nib.load(mask_path)
        mask_data = mask_nii.get_fdata()

        for slice_index in range(mask_data.shape[2]):

            mask_slice = mask_data[:, :, slice_index]

            slice_counts["total"] += 1

            unique_labels = np.unique(mask_slice)

            # Pixel counts
            for class_id in range(4):
                class_pixel_counts[class_id] += np.sum(
                    mask_slice == class_id
                )

            # Slice-level presence
            has_rv = np.any(mask_slice == 1)
            has_myo = np.any(mask_slice == 2)
            has_lv = np.any(mask_slice == 3)

            if not (has_rv or has_myo or has_lv):
                slice_counts["background_only"] += 1
            else:
                slice_counts["all_foreground"] += 1

            if has_rv:
                slice_counts["RV"] += 1

            if has_myo:
                slice_counts["Myocardium"] += 1

            if has_lv:
                slice_counts["LV"] += 1


print("\n===== TRAINING SLICE ANALYSIS =====")

print("Total slices:", slice_counts["total"])

print(
    "Background-only slices:",
    slice_counts["background_only"]
)

print(
    "Foreground-containing slices:",
    slice_counts["all_foreground"]
)

print(
    "Slices containing RV:",
    slice_counts["RV"]
)

print(
    "Slices containing Myocardium:",
    slice_counts["Myocardium"]
)

print(
    "Slices containing LV:",
    slice_counts["LV"]
)


print("\n===== PIXEL DISTRIBUTION =====")

total_pixels = sum(class_pixel_counts.values())

class_names = {
    0: "Background",
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

for class_id, name in class_names.items():

    count = class_pixel_counts[class_id]

    percentage = (
        count / total_pixels
    ) * 100

    print(
        f"{name}: "
        f"{count:,} pixels "
        f"({percentage:.2f}%)"
    )
```

```python
# Step 29A: Focal Tversky Loss

import tensorflow as tf

def focal_tversky_loss(
    y_true,
    y_pred,
    alpha=0.3,
    beta=0.7,
    gamma=0.75,
    smooth=1e-6
):
    # Convert ground truth to one-hot
    y_true = tf.one_hot(
        tf.cast(y_true, tf.int32),
        depth=NUM_CLASSES
    )

    # Flatten
    y_true = tf.reshape(
        y_true,
        [-1, NUM_CLASSES]
    )

    y_pred = tf.reshape(
        y_pred,
        [-1, NUM_CLASSES]
    )

    # True positives
    true_positive = tf.reduce_sum(
        y_true * y_pred,
        axis=0
    )

    # False positives
    false_positive = tf.reduce_sum(
        (1.0 - y_true) * y_pred,
        axis=0
    )

    # False negatives
    false_negative = tf.reduce_sum(
        y_true * (1.0 - y_pred),
        axis=0
    )

    # Tversky index
    tversky = (
        true_positive + smooth
    ) / (
        true_positive
        + alpha * false_positive
        + beta * false_negative
        + smooth
    )

    # Give more importance to RV and myocardium
    class_weights = tf.constant(
        [0.10, 0.40, 0.35, 0.15],
        dtype=tf.float32
    )

    weighted_tversky = tf.reduce_sum(
        tversky * class_weights
    )

    # Focal Tversky
    loss = tf.pow(
        1.0 - weighted_tversky,
        gamma
    )

    return loss


print("Focal Tversky loss created successfully.")
```

```python
# Step 29B: Compile model with Focal Tversky loss

# Restore our previous best model
model.load_weights(improved_checkpoint_path)

optimizer_tversky = tf.keras.optimizers.AdamW(
    learning_rate=0.0001,
    weight_decay=1e-4
)

model.compile(
    optimizer=optimizer_tversky,
    loss=focal_tversky_loss,
    metrics=[dice_coefficient]
)

tversky_checkpoint_path = os.path.join(
    MODEL_DIR,
    "CardioSeg_AI_Tversky_best.keras"
)

tversky_checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
    filepath=tversky_checkpoint_path,
    monitor="val_dice_coefficient",
    mode="max",
    save_best_only=True,
    verbose=1
)

tversky_reduce_lr_callback = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_dice_coefficient",
    mode="max",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

tversky_early_stopping_callback = tf.keras.callbacks.EarlyStopping(
    monitor="val_dice_coefficient",
    mode="max",
    patience=4,
    restore_best_weights=True,
    verbose=1
)

print("Focal Tversky model compiled successfully.")
print("Learning rate: 0.0001")
```

```python
# Step 34: Independent validation Dice check
# No training - prediction/evaluation only

import numpy as np
import tensorflow as tf

print("Running independent validation check...")

all_dice = []
class_dice = {
    0: [],
    1: [],
    2: [],
    3: []
}

for batch_index in range(len(val_generator)):

    X_batch, Y_batch = val_generator[batch_index]

    predictions = model.predict(
        X_batch,
        verbose=0
    )

    pred_labels = np.argmax(
        predictions,
        axis=-1
    )

    for i in range(len(Y_batch)):

        true_mask = Y_batch[i]
        pred_mask = pred_labels[i]

        # Overall foreground + background Dice
        true_flat = tf.one_hot(
            true_mask.astype(np.int32),
            depth=4
        ).numpy()

        pred_flat = tf.one_hot(
            pred_mask.astype(np.int32),
            depth=4
        ).numpy()

        intersection = np.sum(
            true_flat * pred_flat,
            axis=(0, 1)
        )

        denominator = (
            np.sum(true_flat, axis=(0, 1))
            +
            np.sum(pred_flat, axis=(0, 1))
        )

        dice_per_class = np.where(
            denominator > 0,
            (2.0 * intersection) / denominator,
            1.0
        )

        for class_id in range(4):
            class_dice[class_id].append(
                dice_per_class[class_id]
            )

        all_dice.append(
            np.mean(dice_per_class)
        )

print("\n===== INDEPENDENT VALIDATION DICE =====")

print(
    "Overall mean Dice:",
    f"{np.mean(all_dice):.4f}"
)

print(
    "Overall mean Dice:",
    f"{np.mean(all_dice)*100:.2f}%"
)

class_names = {
    0: "Background",
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

print("\n===== CLASS-WISE DICE =====")

for class_id, name in class_names.items():

    score = np.mean(
        class_dice[class_id]
    )

    print(
        f"{name}: "
        f"{score:.4f} "
        f"({score*100:.2f}%)"
    )
```

```python
import numpy as np
import cv2

FINAL_MODEL = saved_model

def predict_mri_slice(image_slice):
    """
    Run CardioSeg AI segmentation on a single MRI slice.
    """

    # 1. Normalize exactly like training
    image_slice = image_slice.astype(np.float32)

    min_value = np.min(image_slice)
    max_value = np.max(image_slice)

    if max_value > min_value:
        image_normalized = (
            (image_slice - min_value)
            / (max_value - min_value)
        )
    else:
        image_normalized = np.zeros_like(image_slice)

    # 2. Resize to model input
    image_resized = cv2.resize(
        image_normalized,
        (128, 128),
        interpolation=cv2.INTER_LINEAR
    )

    # 3. Add channel dimension
    image_input = np.expand_dims(
        image_resized,
        axis=-1
    )

    # 4. Add batch dimension
    image_input = np.expand_dims(
        image_input,
        axis=0
    )

    # 5. Model prediction
    probabilities = FINAL_MODEL.predict(
        image_input,
        verbose=0
    )[0]

    # 6. Convert probabilities to segmentation labels
    predicted_mask = np.argmax(
        probabilities,
        axis=-1
    )

    return (
        image_resized,
        predicted_mask,
        probabilities
    )

print("Final inference function created successfully.")
```

```python
import numpy as np
import nibabel as nib
import os

patient_id = "patient001"
patient_dir = os.path.join(TRAIN_DIR, patient_id)

# ED and ES files
ed_path = os.path.join(
    patient_dir,
    "patient001_frame01.nii.gz"
)

es_path = os.path.join(
    patient_dir,
    "patient001_frame12.nii.gz"
)

# Load images
ed_nii = nib.load(ed_path)
es_nii = nib.load(es_path)

ed_data = ed_nii.get_fdata()
es_data = es_nii.get_fdata()

# Physical voxel dimensions
pixel_spacing_x, pixel_spacing_y, slice_thickness = ed_nii.header.get_zooms()[:3]

pixel_area_mm2 = pixel_spacing_x * pixel_spacing_y
voxel_volume_mm3 = pixel_area_mm2 * slice_thickness

print("Pixel area:", pixel_area_mm2, "mm²")
print("Voxel volume:", voxel_volume_mm3, "mm³")


def segment_volume(image_data):
    """
    Segment all slices and return a 3D predicted mask.
    """

    predicted_volume = np.zeros(
        image_data.shape,
        dtype=np.uint8
    )

    for slice_idx in range(image_data.shape[2]):

        image_slice = image_data[:, :, slice_idx]

        _, predicted_mask, _ = predict_mri_slice(
            image_slice
        )

        # Resize prediction back to original MRI resolution
        predicted_mask_original = cv2.resize(
            predicted_mask.astype(np.uint8),
            (image_data.shape[1], image_data.shape[0]),
            interpolation=cv2.INTER_NEAREST
        )

        predicted_volume[:, :, slice_idx] = predicted_mask_original

    return predicted_volume


# Generate ED and ES segmentations
print("\nGenerating ED segmentation...")
ed_prediction = segment_volume(ed_data)

print("Generating ES segmentation...")
es_prediction = segment_volume(es_data)

print("\nED predicted labels:", np.unique(ed_prediction))
print("ES predicted labels:", np.unique(es_prediction))


# LV = class 3
lv_ed_pixels = np.sum(ed_prediction == 3)
lv_es_pixels = np.sum(es_prediction == 3)

# Convert voxel counts to volume
lv_ed_volume_mm3 = lv_ed_pixels * voxel_volume_mm3
lv_es_volume_mm3 = lv_es_pixels * voxel_volume_mm3

# Convert mm³ to mL
lv_ed_volume_ml = lv_ed_volume_mm3 / 1000
lv_es_volume_ml = lv_es_volume_mm3 / 1000

# LVEF
if lv_ed_volume_ml > 0:
    lvef = (
        (lv_ed_volume_ml - lv_es_volume_ml)
        / lv_ed_volume_ml
    ) * 100
else:
    lvef = 0.0

print("\n================================")
print("CARDIOSEG AI CARDIAC MEASUREMENTS")
print("================================")
print(f"LV EDV : {lv_ed_volume_ml:.2f} mL")
print(f"LV ESV : {lv_es_volume_ml:.2f} mL")
print(f"LVEF   : {lvef:.2f}%")
print("================================")
```

```python
# ==========================================
# RV AND MYOCARDIUM VOLUME CALCULATION
# ==========================================

# Class IDs
RV_CLASS = 1
MYOCARDIUM_CLASS = 2
LV_CLASS = 3

def calculate_class_volume(prediction, class_id):
    """
    Calculate segmented volume in mL.
    """
    voxel_count = np.sum(prediction == class_id)

    volume_mm3 = voxel_count * voxel_volume_mm3

    volume_ml = volume_mm3 / 1000.0

    return volume_ml


# -----------------------------
# RV volumes
# -----------------------------

rv_ed_volume_ml = calculate_class_volume(
    ed_prediction,
    RV_CLASS
)

rv_es_volume_ml = calculate_class_volume(
    es_prediction,
    RV_CLASS
)


# -----------------------------
# Myocardium volumes
# -----------------------------

myo_ed_volume_ml = calculate_class_volume(
    ed_prediction,
    MYOCARDIUM_CLASS
)

myo_es_volume_ml = calculate_class_volume(
    es_prediction,
    MYOCARDIUM_CLASS
)


# -----------------------------
# LV volumes
# -----------------------------

lv_ed_volume_ml = calculate_class_volume(
    ed_prediction,
    LV_CLASS
)

lv_es_volume_ml = calculate_class_volume(
    es_prediction,
    LV_CLASS
)


# -----------------------------
# Print results
# -----------------------------

print("==========================================")
print("CARDIOSEG AI — CARDIAC VOLUME ANALYSIS")
print("==========================================")

print("\nLEFT VENTRICLE")
print(f"LV EDV : {lv_ed_volume_ml:.2f} mL")
print(f"LV ESV : {lv_es_volume_ml:.2f} mL")
print(f"LVEF   : {lvef:.2f}%")

print("\nRIGHT VENTRICLE")
print(f"RV EDV : {rv_ed_volume_ml:.2f} mL")
print(f"RV ESV : {rv_es_volume_ml:.2f} mL")

print("\nMYOCARDIUM")
print(f"Myo ED volume : {myo_ed_volume_ml:.2f} mL")
print(f"Myo ES volume : {myo_es_volume_ml:.2f} mL")

print("\n==========================================")
```

```python
import matplotlib.pyplot as plt

# Select the middle slice
mid_slice = ed_prediction.shape[2] // 2

# ED and ES MRI slices
ed_slice = ed_data[:, :, mid_slice]
es_slice = es_data[:, :, mid_slice]

# Predicted masks
ed_mask = ed_prediction[:, :, mid_slice]
es_mask = es_prediction[:, :, mid_slice]

# Normalize MRI images for display
def normalize_for_display(image):
    image = image.astype(np.float32)
    min_val = image.min()
    max_val = image.max()

    if max_val > min_val:
        return (image - min_val) / (max_val - min_val)

    return np.zeros_like(image)


ed_display = normalize_for_display(ed_slice)
es_display = normalize_for_display(es_slice)

# Create figure
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# ED MRI
axes[0, 0].imshow(ed_display, cmap="gray")
axes[0, 0].set_title("End-Diastole (ED) - MRI")
axes[0, 0].axis("off")

# ED segmentation
axes[0, 1].imshow(ed_display, cmap="gray")
axes[0, 1].imshow(
    np.ma.masked_where(ed_mask == 0, ed_mask),
    alpha=0.5,
    cmap="jet",
    vmin=1,
    vmax=3
)
axes[0, 1].set_title("ED - CardioSeg AI Segmentation")
axes[0, 1].axis("off")

# ES MRI
axes[1, 0].imshow(es_display, cmap="gray")
axes[1, 0].set_title("End-Systole (ES) - MRI")
axes[1, 0].axis("off")

# ES segmentation
axes[1, 1].imshow(es_display, cmap="gray")
axes[1, 1].imshow(
    np.ma.masked_where(es_mask == 0, es_mask),
    alpha=0.5,
    cmap="jet",
    vmin=1,
    vmax=3
)
axes[1, 1].set_title("ES - CardioSeg AI Segmentation")
axes[1, 1].axis("off")

plt.tight_layout()

# Save result
ed_es_path = os.path.join(
    RESULTS_DIR,
    "patient001_ED_ES_segmentation.png"
)

fig.savefig(
    ed_es_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("ED/ES visualization saved:")
print(ed_es_path)
```

```python
import os
import shutil
import tensorflow as tf
import nibabel as nib

from PIL import Image
from fastapi import FastAPI, UploadFile, File, HTTPException

MODEL_PATH = "/content/drive/MyDrive/CardioSeg_AI/models/CardioSeg_AI_Boundary_best.keras"
UPLOAD_DIR = "/content/drive/MyDrive/CardioSeg_AI/backend/uploads"

os.makedirs(UPLOAD_DIR, exist_ok=True)

app = FastAPI(
    title="CardioSeg AI API",
    description="AI-powered Cardiac MRI Segmentation and Analysis",
    version="1.0.0"
)

# Load trained model
CARDIOSEG_MODEL = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

@app.get("/")
def root():
    return {
        "message": "CardioSeg AI API is running",
        "status": "online"
    }

@app.get("/health")
def health():
    return {
        "status": "healthy",
        "service": "CardioSeg AI"
    }

@app.get("/model-status")
def model_status():
    return {
        "model": "CardioSeg AI",
        "architecture": "2D U-Net",
        "input_size": "128x128x1",
        "output_classes": 4,
        "classes": {
            "0": "Background",
            "1": "RV",
            "2": "Myocardium",
            "3": "LV"
        },
        "parameters": CARDIOSEG_MODEL.count_params(),
        "status": "loaded"
    }

@app.post("/upload-mri")
async def upload_mri(file: UploadFile = File(...)):

    filename = file.filename.lower()

    allowed_nifti = filename.endswith(".nii") or filename.endswith(".nii.gz")
    allowed_image = (
        filename.endswith(".png")
        or filename.endswith(".jpg")
        or filename.endswith(".jpeg")
    )

    if not (allowed_nifti or allowed_image):
        raise HTTPException(
            status_code=400,
            detail="Supported files: .nii, .nii.gz, .png, .jpg, .jpeg"
        )

    save_path = os.path.join(UPLOAD_DIR, file.filename)

    with open(save_path, "wb") as buffer:
        shutil.copyfileobj(file.file, buffer)

    try:

        # -------------------------------
        # NIfTI MRI
        # -------------------------------
        if allowed_nifti:

            mri = nib.load(save_path)

            return {
                "status": "uploaded",
                "file_type": "NIfTI MRI",
                "filename": file.filename,
                "path": save_path,
                "shape": list(mri.shape),
                "voxel_spacing_mm": list(mri.header.get_zooms()),
                "analysis_type": "Full cardiac MRI analysis available",
                "message": "NIfTI MRI uploaded and successfully read."
            }

        # -------------------------------
        # PNG / JPG / JPEG
        # -------------------------------
        if allowed_image:

            image = Image.open(save_path)

            return {
                "status": "uploaded",
                "file_type": "MRI image slice",
                "filename": file.filename,
                "path": save_path,
                "image_size": list(image.size),
                "image_mode": image.mode,
                "analysis_type": "2D slice segmentation",
                "message": (
                    "MRI image uploaded successfully. "
                    "2D segmentation can be performed."
                )
            }

    except Exception as e:

        if os.path.exists(save_path):
            os.remove(save_path)

        raise HTTPException(
            status_code=400,
            detail=f"Unable to read uploaded file: {str(e)}"
        )
```

```python
import os
import nibabel as nib
import numpy as np
import cv2

# Existing ACDC MRI
TEST_NII = (
    "/content/drive/MyDrive/CardioSeg_AI/"
    "dataset/ACDC/database/training/"
    "patient001/patient001_frame01.nii.gz"
)

UPLOAD_DIR = "/content/drive/MyDrive/CardioSeg_AI/backend/uploads"
os.makedirs(UPLOAD_DIR, exist_ok=True)

# Load MRI
nii = nib.load(TEST_NII)
mri = nii.get_fdata()

# Select middle slice
slice_index = mri.shape[2] // 2
image_slice = mri[:, :, slice_index].astype(np.float32)

# Normalize to 0-255
min_val = image_slice.min()
max_val = image_slice.max()

if max_val > min_val:
    image_norm = (image_slice - min_val) / (max_val - min_val)
else:
    image_norm = np.zeros_like(image_slice)

image_png = (image_norm * 255).astype(np.uint8)

# Save PNG
TEST_PNG = os.path.join(UPLOAD_DIR, "patient001_test_slice.png")
cv2.imwrite(TEST_PNG, image_png)

print("======================================")
print("Test MRI Files")
print("======================================")
print("NIfTI:", TEST_NII)
print("NIfTI shape:", mri.shape)
print("Test slice:", slice_index)
print("PNG:", TEST_PNG)
print("PNG shape:", image_png.shape)
print("======================================")
```

```python
from pathlib import Path

BACKEND_DIR = Path("/content/drive/MyDrive/CardioSeg_AI/backend")
MAIN_FILE = BACKEND_DIR / "main.py"

main_code = r'''
import os
import shutil
import tensorflow as tf
import nibabel as nib

from PIL import Image
from fastapi import FastAPI, UploadFile, File, HTTPException

MODEL_PATH = "/content/drive/MyDrive/CardioSeg_AI/models/CardioSeg_AI_Boundary_best.keras"
UPLOAD_DIR = "/content/drive/MyDrive/CardioSeg_AI/backend/uploads"

os.makedirs(UPLOAD_DIR, exist_ok=True)

app = FastAPI(
    title="CardioSeg AI API",
    description="AI-powered Cardiac MRI Segmentation and Analysis",
    version="1.0.0"
)

# Load trained CardioSeg AI model
CARDIOSEG_MODEL = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)


@app.get("/")
def root():
    return {
        "message": "CardioSeg AI API is running",
        "status": "online"
    }


@app.get("/health")
def health():
    return {
        "status": "healthy",
        "service": "CardioSeg AI"
    }


@app.get("/model-status")
def model_status():
    return {
        "model": "CardioSeg AI",
        "architecture": "2D U-Net",
        "input_size": "128x128x1",
        "output_classes": 4,
        "classes": {
            "0": "Background",
            "1": "RV",
            "2": "Myocardium",
            "3": "LV"
        },
        "parameters": CARDIOSEG_MODEL.count_params(),
        "status": "loaded"
    }


@app.post("/upload-mri")
async def upload_mri(file: UploadFile = File(...)):

    filename = file.filename.lower()

    allowed_nifti = (
        filename.endswith(".nii") or
        filename.endswith(".nii.gz")
    )

    allowed_image = (
        filename.endswith(".png") or
        filename.endswith(".jpg") or
        filename.endswith(".jpeg")
    )

    if not (allowed_nifti or allowed_image):
        raise HTTPException(
            status_code=400,
            detail="Supported files: .nii, .nii.gz, .png, .jpg, .jpeg"
        )

    save_path = os.path.join(
        UPLOAD_DIR,
        file.filename
    )

    with open(save_path, "wb") as buffer:
        shutil.copyfileobj(file.file, buffer)

    try:

        # -----------------------------
        # NIfTI MRI
        # -----------------------------
        if allowed_nifti:

            mri = nib.load(save_path)

            return {
                "status": "uploaded",
                "file_type": "NIfTI MRI",
                "filename": file.filename,
                "path": save_path,
                "shape": list(mri.shape),
                "voxel_spacing_mm": list(mri.header.get_zooms()),
                "analysis_type": "Full cardiac MRI analysis available",
                "message": "NIfTI MRI uploaded and successfully read."
            }

        # -----------------------------
        # PNG / JPG / JPEG
        # -----------------------------
        if allowed_image:

            image = Image.open(save_path)

            return {
                "status": "uploaded",
                "file_type": "MRI image slice",
                "filename": file.filename,
                "path": save_path,
                "image_size": list(image.size),
                "image_mode": image.mode,
                "analysis_type": "2D slice segmentation",
                "message": (
                    "MRI image uploaded successfully. "
                    "2D segmentation can be performed."
                )
            }

    except Exception as e:

        if os.path.exists(save_path):
            os.remove(save_path)

        raise HTTPException(
            status_code=400,
            detail=f"Unable to read uploaded file: {str(e)}"
        )
'''

MAIN_FILE.write_text(main_code)

print("======================================")
print("MAIN.PY UPDATED")
print("======================================")
print("File:", MAIN_FILE)
print("Exists:", MAIN_FILE.exists())
print("Size:", MAIN_FILE.stat().st_size, "bytes")

# Verify the important route is physically present
content = MAIN_FILE.read_text()

if '@app.post("/upload-mri")' in content:
    print("✅ /upload-mri is present in main.py")
else:
    print("❌ /upload-mri is NOT present")
```

```python
from pathlib import Path

BACKEND_DIR = Path("/content/drive/MyDrive/CardioSeg_AI/backend")
MAIN_FILE = BACKEND_DIR / "main.py"

main_code = r'''
import os
import shutil
import numpy as np
import cv2
import tensorflow as tf
import nibabel as nib

from PIL import Image
from fastapi import FastAPI, UploadFile, File, HTTPException

MODEL_PATH = "/content/drive/MyDrive/CardioSeg_AI/models/CardioSeg_AI_Boundary_best.keras"
UPLOAD_DIR = "/content/drive/MyDrive/CardioSeg_AI/backend/uploads"
RESULT_DIR = "/content/drive/MyDrive/CardioSeg_AI/backend/results"

os.makedirs(UPLOAD_DIR, exist_ok=True)
os.makedirs(RESULT_DIR, exist_ok=True)

app = FastAPI(
    title="CardioSeg AI API",
    description="AI-powered Cardiac MRI Segmentation and Analysis",
    version="1.0.0"
)

# Load trained CardioSeg AI model
CARDIOSEG_MODEL = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)


@app.get("/")
def root():
    return {
        "message": "CardioSeg AI API is running",
        "status": "online"
    }


@app.get("/health")
def health():
    return {
        "status": "healthy",
        "service": "CardioSeg AI"
    }


@app.get("/model-status")
def model_status():
    return {
        "model": "CardioSeg AI",
        "architecture": "2D U-Net",
        "input_size": "128x128x1",
        "output_classes": 4,
        "classes": {
            "0": "Background",
            "1": "RV",
            "2": "Myocardium",
            "3": "LV"
        },
        "parameters": CARDIOSEG_MODEL.count_params(),
        "status": "loaded"
    }


@app.post("/upload-mri")
async def upload_mri(file: UploadFile = File(...)):

    filename = file.filename.lower()

    allowed_nifti = (
        filename.endswith(".nii") or
        filename.endswith(".nii.gz")
    )

    allowed_image = (
        filename.endswith(".png") or
        filename.endswith(".jpg") or
        filename.endswith(".jpeg")
    )

    if not (allowed_nifti or allowed_image):
        raise HTTPException(
            status_code=400,
            detail="Supported files: .nii, .nii.gz, .png, .jpg, .jpeg"
        )

    save_path = os.path.join(
        UPLOAD_DIR,
        file.filename
    )

    with open(save_path, "wb") as buffer:
        shutil.copyfileobj(file.file, buffer)

    try:

        if allowed_nifti:

            mri = nib.load(save_path)

            return {
                "status": "uploaded",
                "file_type": "NIfTI MRI",
                "filename": file.filename,
                "path": save_path,
                "shape": list(mri.shape),
                "voxel_spacing_mm": list(mri.header.get_zooms()),
                "analysis_type": "Full cardiac MRI analysis available",
                "message": "NIfTI MRI uploaded and successfully read."
            }

        if allowed_image:

            image = Image.open(save_path)

            return {
                "status": "uploaded",
                "file_type": "MRI image slice",
                "filename": file.filename,
                "path": save_path,
                "image_size": list(image.size),
                "image_mode": image.mode,
                "analysis_type": "2D slice segmentation",
                "message": (
                    "MRI image uploaded successfully. "
                    "2D segmentation can be performed."
                )
            }

    except Exception as e:

        if os.path.exists(save_path):
            os.remove(save_path)

        raise HTTPException(
            status_code=400,
            detail=f"Unable to read uploaded file: {str(e)}"
        )


@app.post("/predict")
async def predict_mri(file: UploadFile = File(...)):

    filename = file.filename.lower()

    allowed_image = (
        filename.endswith(".png")
        or filename.endswith(".jpg")
        or filename.endswith(".jpeg")
    )

    if not allowed_image:
        raise HTTPException(
            status_code=400,
            detail="For this endpoint, upload a PNG, JPG, or JPEG MRI slice."
        )

    save_path = os.path.join(
        UPLOAD_DIR,
        file.filename
    )

    with open(save_path, "wb") as buffer:
        shutil.copyfileobj(file.file, buffer)

    try:

        # -----------------------------
        # Read grayscale MRI slice
        # -----------------------------
        image = Image.open(save_path).convert("L")
        image_array = np.array(image).astype(np.float32)

        # -----------------------------
        # Same normalization as training
        # -----------------------------
        min_value = np.min(image_array)
        max_value = np.max(image_array)

        if max_value > min_value:
            image_normalized = (
                (image_array - min_value) /
                (max_value - min_value)
            )
        else:
            image_normalized = np.zeros_like(image_array)

        # -----------------------------
        # Resize to model input size
        # -----------------------------
        image_resized = cv2.resize(
            image_normalized,
            (128, 128),
            interpolation=cv2.INTER_LINEAR
        )

        # -----------------------------
        # Model input
        # -----------------------------
        model_input = np.expand_dims(
            image_resized,
            axis=(0, -1)
        )

        # -----------------------------
        # Prediction
        # -----------------------------
        probabilities = CARDIOSEG_MODEL.predict(
            model_input,
            verbose=0
        )[0]

        predicted_mask = np.argmax(
            probabilities,
            axis=-1
        ).astype(np.uint8)

        # -----------------------------
        # Class pixel counts
        # -----------------------------
        background_pixels = int(
            np.sum(predicted_mask == 0)
        )

        rv_pixels = int(
            np.sum(predicted_mask == 1)
        )

        myocardium_pixels = int(
            np.sum(predicted_mask == 2)
        )

        lv_pixels = int(
            np.sum(predicted_mask == 3)
        )

        # -----------------------------
        # Save predicted mask
        # -----------------------------
        base_name = os.path.splitext(
            os.path.basename(file.filename)
        )[0]

        mask_filename = (
            base_name +
            "_predicted_mask.png"
        )

        mask_path = os.path.join(
            RESULT_DIR,
            mask_filename
        )

        Image.fromarray(
            predicted_mask
        ).save(mask_path)

        # -----------------------------
        # Confidence
        # -----------------------------
        mean_confidence = float(
            np.mean(
                np.max(
                    probabilities,
                    axis=-1
                )
            )
        )

        return {
            "status": "success",
            "filename": file.filename,
            "model": "CardioSeg AI",
            "architecture": "2D U-Net",
            "input_size": [128, 128],
            "prediction_classes": {
                "0": "Background",
                "1": "RV",
                "2": "Myocardium",
                "3": "LV"
            },
            "pixel_counts": {
                "background": background_pixels,
                "RV": rv_pixels,
                "myocardium": myocardium_pixels,
                "LV": lv_pixels
            },
            "mean_prediction_confidence": mean_confidence,
            "predicted_mask_path": mask_path,
            "dice_available": False,
            "message": (
                "MRI slice segmented successfully. "
                "Dice and IoU are not reported because "
                "ground-truth segmentation was not provided."
            )
        }

    except Exception as e:

        if os.path.exists(save_path):
            os.remove(save_path)

        raise HTTPException(
            status_code=500,
            detail=f"Segmentation failed: {str(e)}"
        )
'''

MAIN_FILE.write_text(main_code)

print("======================================")
print("PREDICTION API UPDATED")
print("======================================")
print("File:", MAIN_FILE)
print("Exists:", MAIN_FILE.exists())

content = MAIN_FILE.read_text()

if '@app.post("/predict")' in content:
    print("✅ /predict endpoint added")
else:
    print("❌ /predict endpoint NOT found")

print("======================================")
```

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# ---------------------------------------
# File paths
# ---------------------------------------
MRI_PATH = "/content/drive/MyDrive/CardioSeg_AI/backend/uploads/patient001_test_slice.png"

MASK_PATH = "/content/drive/MyDrive/CardioSeg_AI/backend/results/patient001_test_slice_predicted_mask.png"

RESULT_DIR = "/content/drive/MyDrive/CardioSeg_AI/backend/results"

OVERLAY_PATH = os.path.join(
    RESULT_DIR,
    "patient001_test_slice_segmentation_overlay.png"
)

# ---------------------------------------
# Load images
# ---------------------------------------
mri = np.array(
    Image.open(MRI_PATH).convert("L")
)

mask = np.array(
    Image.open(MASK_PATH)
)

# ---------------------------------------
# Resize MRI to mask size
# ---------------------------------------
mri_resized = cv2.resize(
    mri,
    (mask.shape[1], mask.shape[0]),
    interpolation=cv2.INTER_LINEAR
)

# Normalize MRI for display
mri_normalized = cv2.normalize(
    mri_resized,
    None,
    0,
    255,
    cv2.NORM_MINMAX
).astype(np.uint8)

# Convert grayscale MRI to RGB
overlay = cv2.cvtColor(
    mri_normalized,
    cv2.COLOR_GRAY2RGB
)

# ---------------------------------------
# Create segmentation colors
#
# Background     = black
# RV             = red
# Myocardium     = green
# LV             = blue
# ---------------------------------------
segmentation = np.zeros_like(overlay)

segmentation[mask == 1] = [255, 0, 0]      # RV
segmentation[mask == 2] = [0, 255, 0]      # Myocardium
segmentation[mask == 3] = [0, 0, 255]      # LV

# ---------------------------------------
# Blend MRI + segmentation
# ---------------------------------------
alpha = 0.45

colored_pixels = mask > 0

overlay[colored_pixels] = (
    (1 - alpha) * overlay[colored_pixels]
    + alpha * segmentation[colored_pixels]
).astype(np.uint8)

# ---------------------------------------
# Save overlay
# ---------------------------------------
Image.fromarray(overlay).save(
    OVERLAY_PATH
)

print("======================================")
print("SEGMENTATION VISUALIZATION")
print("======================================")

print("Original MRI:")
print(MRI_PATH)

print("\nPredicted mask:")
print(MASK_PATH)

print("\nOverlay:")
print(OVERLAY_PATH)

print("\nMask shape:", mask.shape)

print("\nClass pixel counts:")
print("Background :", int(np.sum(mask == 0)))
print("RV         :", int(np.sum(mask == 1)))
print("Myocardium :", int(np.sum(mask == 2)))
print("LV         :", int(np.sum(mask == 3)))

print("\n✅ Overlay created successfully!")

print("======================================")

# ---------------------------------------
# Display results
# ---------------------------------------
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(mri_normalized, cmap="gray")
plt.title("Original MRI")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(mask, cmap="viridis")
plt.title("Predicted Segmentation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(overlay)
plt.title("MRI + Segmentation Overlay")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path

BACKEND_DIR = Path("/content/drive/MyDrive/CardioSeg_AI/backend")
MAIN_FILE = BACKEND_DIR / "main.py"

content = MAIN_FILE.read_text()

# Add overlay generation after the predicted mask is saved.
old_block = '''        Image.fromarray(
            predicted_mask
        ).save(mask_path)

        # -----------------------------
        # Confidence
        # -----------------------------'''

new_block = '''        Image.fromarray(
            predicted_mask
        ).save(mask_path)

        # -----------------------------
        # Create segmentation overlay
        # -----------------------------
        mri_display = cv2.normalize(
            image_array,
            None,
            0,
            255,
            cv2.NORM_MINMAX
        ).astype(np.uint8)

        mri_display = cv2.resize(
            mri_display,
            (128, 128),
            interpolation=cv2.INTER_LINEAR
        )

        overlay = cv2.cvtColor(
            mri_display,
            cv2.COLOR_GRAY2RGB
        )

        segmentation = np.zeros_like(overlay)

        # RV = red
        segmentation[predicted_mask == 1] = [255, 0, 0]

        # Myocardium = green
        segmentation[predicted_mask == 2] = [0, 255, 0]

        # LV = blue
        segmentation[predicted_mask == 3] = [0, 0, 255]

        foreground = predicted_mask > 0

        alpha = 0.45

        overlay[foreground] = (
            (1 - alpha) * overlay[foreground]
            + alpha * segmentation[foreground]
        ).astype(np.uint8)

        overlay_filename = (
            base_name +
            "_segmentation_overlay.png"
        )

        overlay_path = os.path.join(
            RESULT_DIR,
            overlay_filename
        )

        Image.fromarray(
            overlay
        ).save(overlay_path)

        # -----------------------------
        # Confidence
        # -----------------------------'''

if old_block not in content:
    print("❌ Expected code block was not found.")
    print("No changes were made.")

else:
    content = content.replace(old_block, new_block)

    # Add overlay path to response
    old_response = '''            "predicted_mask_path": mask_path,
            "dice_available": False,'''

    new_response = '''            "predicted_mask_path": mask_path,
            "overlay_path": overlay_path,
            "dice_available": False,'''

    content = content.replace(
        old_response,
        new_response
    )

    MAIN_FILE.write_text(content)

    print("======================================")
    print("OVERLAY API UPDATE")
    print("======================================")
    print("✅ Overlay generation added to /predict")
    print("✅ overlay_path added to API response")
    print("======================================")
```

```python
import os
import numpy as np
import nibabel as nib
import cv2

NIFTI_RESULT_DIR = "/content/drive/MyDrive/CardioSeg_AI/backend/results/nifti"
os.makedirs(NIFTI_RESULT_DIR, exist_ok=True)


def predict_nifti_volume(nifti_path, output_path):
    """
    Segment every 2D slice of a 3D cardiac MRI volume
    using the trained CardioSeg AI 2D U-Net.
    """

    # Load NIfTI
    nii = nib.load(nifti_path)
    volume = nii.get_fdata().astype(np.float32)

    print("Original volume shape:", volume.shape)

    # Handle 3D volume
    if volume.ndim != 3:
        raise ValueError(
            f"Expected a 3D MRI volume, but got shape {volume.shape}"
        )

    # Output mask has same spatial shape as input
    predicted_volume = np.zeros(
        volume.shape,
        dtype=np.uint8
    )

    # Process every slice
    for slice_index in range(volume.shape[2]):

        image_slice = volume[:, :, slice_index]

        # Normalize
        min_value = np.min(image_slice)
        max_value = np.max(image_slice)

        if max_value > min_value:
            normalized = (
                (image_slice - min_value)
                / (max_value - min_value)
            )
        else:
            normalized = np.zeros_like(image_slice)

        # Resize to model input
        resized = cv2.resize(
            normalized,
            (128, 128),
            interpolation=cv2.INTER_LINEAR
        )

        # Model input
        model_input = np.expand_dims(
            resized,
            axis=(0, -1)
        )

        # Prediction
        probabilities = CARDIOSEG_MODEL.predict(
            model_input,
            verbose=0
        )[0]

        predicted_small = np.argmax(
            probabilities,
            axis=-1
        ).astype(np.uint8)

        # Resize prediction back to original slice size
        predicted_original = cv2.resize(
            predicted_small,
            (volume.shape[1], volume.shape[0]),
            interpolation=cv2.INTER_NEAREST
        )

        predicted_volume[:, :, slice_index] = predicted_original

        if (slice_index + 1) % 5 == 0 or slice_index == 0:
            print(
                f"Processed slice "
                f"{slice_index + 1}/{volume.shape[2]}"
            )

    # Save NIfTI prediction
    prediction_nii = nib.Nifti1Image(
        predicted_volume,
        nii.affine,
        nii.header
    )

    nib.save(
        prediction_nii,
        output_path
    )

    print("\n======================================")
    print("NIFTI SEGMENTATION COMPLETE")
    print("======================================")
    print("Input shape :", volume.shape)
    print("Output shape:", predicted_volume.shape)
    print("Output file :", output_path)

    print("\nClass pixel/voxel counts:")
    print("Background :", np.sum(predicted_volume == 0))
    print("RV          :", np.sum(predicted_volume == 1))
    print("Myocardium  :", np.sum(predicted_volume == 2))
    print("LV          :", np.sum(predicted_volume == 3))

    print("======================================")

    return predicted_volume, nii
```

```python
# ============================================
# STEP 72E - CARDIAC VOLUME CALCULATION
# ============================================

# Get voxel spacing from the original MRI
ed_spacing = ed_nii.header.get_zooms()[:3]

print("Voxel spacing (mm):", ed_spacing)

# Volume of one voxel in mm³
voxel_volume_mm3 = (
    ed_spacing[0]
    * ed_spacing[1]
    * ed_spacing[2]
)

# Convert mm³ to mL
# 1000 mm³ = 1 mL
voxel_volume_ml = voxel_volume_mm3 / 1000.0

print("Voxel volume (mm³):", voxel_volume_mm3)
print("Voxel volume (mL):", voxel_volume_ml)


# ============================================
# ED VOLUMES
# ============================================

lv_ed_voxels = np.sum(predicted_ed == 3)
rv_ed_voxels = np.sum(predicted_ed == 1)
myo_ed_voxels = np.sum(predicted_ed == 2)

lv_edv = lv_ed_voxels * voxel_volume_ml
rv_edv = rv_ed_voxels * voxel_volume_ml
myo_ed_volume = myo_ed_voxels * voxel_volume_ml


# ============================================
# ES VOLUMES
# ============================================

lv_es_voxels = np.sum(predicted_es == 3)
rv_es_voxels = np.sum(predicted_es == 1)
myo_es_voxels = np.sum(predicted_es == 2)

lv_esv = lv_es_voxels * voxel_volume_ml
rv_esv = rv_es_voxels * voxel_volume_ml
myo_es_volume = myo_es_voxels * voxel_volume_ml


# ============================================
# LVEF
# ============================================

if lv_edv > 0:
    lvef = ((lv_edv - lv_esv) / lv_edv) * 100
else:
    lvef = 0.0


# ============================================
# RESULTS
# ============================================

print("\n======================================")
print("CARDIAC MEASUREMENTS")
print("======================================")

print(f"LV EDV          : {lv_edv:.2f} mL")
print(f"LV ESV          : {lv_esv:.2f} mL")
print(f"LVEF            : {lvef:.2f} %")

print(f"RV EDV          : {rv_edv:.2f} mL")
print(f"RV ESV          : {rv_esv:.2f} mL")

print(f"Myocardium ED   : {myo_ed_volume:.2f} mL")
print(f"Myocardium ES   : {myo_es_volume:.2f} mL")

print("\nVoxel counts:")
print(f"LV ED voxels    : {lv_ed_voxels}")
print(f"LV ES voxels    : {lv_es_voxels}")
print(f"RV ED voxels    : {rv_ed_voxels}")
print(f"RV ES voxels    : {rv_es_voxels}")
print(f"Myo ED voxels   : {myo_ed_voxels}")
print(f"Myo ES voxels   : {myo_es_voxels}")

print("======================================")
```

```python
# ============================================
# STEP 72F - ED SEGMENTATION EVALUATION
# ============================================

GT_ED_PATH = "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/training/patient001/patient001_frame01_gt.nii.gz"

gt_ed_nii = nib.load(GT_ED_PATH)
ground_truth_ed = gt_ed_nii.get_fdata().astype(np.uint8)

print("Ground truth shape :", ground_truth_ed.shape)
print("Prediction shape   :", predicted_ed.shape)

print("GT classes         :", np.unique(ground_truth_ed))
print("Prediction classes :", np.unique(predicted_ed))


def dice_score_binary(pred, gt):
    pred = pred.astype(bool)
    gt = gt.astype(bool)

    intersection = np.sum(pred & gt)

    denominator = np.sum(pred) + np.sum(gt)

    if denominator == 0:
        return 1.0

    return (2.0 * intersection) / denominator


def iou_score_binary(pred, gt):
    pred = pred.astype(bool)
    gt = gt.astype(bool)

    intersection = np.sum(pred & gt)
    union = np.sum(pred | gt)

    if union == 0:
        return 1.0

    return intersection / union


print("\n======================================")
print("ED SEGMENTATION QUALITY")
print("======================================")

class_names = {
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

for class_id, class_name in class_names.items():

    pred_class = predicted_ed == class_id
    gt_class = ground_truth_ed == class_id

    dice = dice_score_binary(
        pred_class,
        gt_class
    )

    iou = iou_score_binary(
        pred_class,
        gt_class
    )

    print(
        f"{class_name:12s} | "
        f"Dice: {dice:.4f} | "
        f"IoU: {iou:.4f}"
    )

print("======================================")
```

```python
# ============================================
# STEP 72G - ES SEGMENTATION EVALUATION
# ============================================

GT_ES_PATH = "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/training/patient001/patient001_frame12_gt.nii.gz"

gt_es_nii = nib.load(GT_ES_PATH)
ground_truth_es = gt_es_nii.get_fdata().astype(np.uint8)

print("Ground truth shape :", ground_truth_es.shape)
print("Prediction shape   :", predicted_es.shape)

print("GT classes         :", np.unique(ground_truth_es))
print("Prediction classes :", np.unique(predicted_es))


print("\n======================================")
print("ES SEGMENTATION QUALITY")
print("======================================")

for class_id, class_name in class_names.items():

    pred_class = predicted_es == class_id
    gt_class = ground_truth_es == class_id

    dice = dice_score_binary(
        pred_class,
        gt_class
    )

    iou = iou_score_binary(
        pred_class,
        gt_class
    )

    print(
        f"{class_name:12s} | "
        f"Dice: {dice:.4f} | "
        f"IoU: {iou:.4f}"
    )

print("======================================")
```

```python
# ============================================
# STEP 72H - COMPLETE PATIENT RESULT SUMMARY
# ============================================

ed_metrics = {}
es_metrics = {}

for class_id, class_name in class_names.items():

    # ED
    ed_metrics[class_name] = {
        "dice": dice_score_binary(
            predicted_ed == class_id,
            ground_truth_ed == class_id
        ),
        "iou": iou_score_binary(
            predicted_ed == class_id,
            ground_truth_ed == class_id
        )
    }

    # ES
    es_metrics[class_name] = {
        "dice": dice_score_binary(
            predicted_es == class_id,
            ground_truth_es == class_id
        ),
        "iou": iou_score_binary(
            predicted_es == class_id,
            ground_truth_es == class_id
        )
    }


patient_result = {
    "patient_id": "patient001",

    "phases": {
        "ED": {
            "frame": 1,
            "segmentation_file": ED_OUTPUT
        },
        "ES": {
            "frame": 12,
            "segmentation_file": ES_OUTPUT
        }
    },

    "voxel_spacing_mm": [
        float(ed_spacing[0]),
        float(ed_spacing[1]),
        float(ed_spacing[2])
    ],

    "cardiac_measurements": {
        "LV_EDV_ml": float(lv_edv),
        "LV_ESV_ml": float(lv_esv),
        "LVEF_percent": float(lvef),
        "RV_EDV_ml": float(rv_edv),
        "RV_ESV_ml": float(rv_esv),
        "Myocardium_ED_ml": float(myo_ed_volume),
        "Myocardium_ES_ml": float(myo_es_volume)
    },

    "validation_metrics": {
        "ED": ed_metrics,
        "ES": es_metrics
    }
}

print("======================================")
print("CARDIOSEG AI - COMPLETE RESULT")
print("======================================")

print(patient_result)
print("======================================")
```

```python
import json
import numpy as np
import os

# Convert NumPy values to normal Python values
def convert_numpy(obj):
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    if isinstance(obj, dict):
        return {
            key: convert_numpy(value)
            for key, value in obj.items()
        }
    if isinstance(obj, list):
        return [
            convert_numpy(value)
            for value in obj
        ]
    return obj


patient_result_clean = convert_numpy(patient_result)

JSON_OUTPUT = "/content/drive/MyDrive/CardioSeg_AI/backend/results/nifti/patient001_complete_result.json"

with open(JSON_OUTPUT, "w") as f:
    json.dump(
        patient_result_clean,
        f,
        indent=4
    )

print("======================================")
print("RESULT JSON SAVED")
print("======================================")
print("File:", JSON_OUTPUT)
print("Exists:", os.path.exists(JSON_OUTPUT))

if os.path.exists(JSON_OUTPUT):
    print(
        "Size:",
        os.path.getsize(JSON_OUTPUT),
        "bytes"
    )

print("======================================")
```

```python
MAIN_PY = "/content/drive/MyDrive/CardioSeg_AI/backend/main.py"

with open(MAIN_PY, "r") as f:
    content = f.read()

start_marker = '@app.post("/analyze-nifti")'

if start_marker not in content:
    print("❌ /analyze-nifti endpoint was not found.")
else:

    start = content.index(start_marker)

    # Since this is currently the last endpoint,
    # replace everything from /analyze-nifti onward.
    content_before_endpoint = content[:start]

    NEW_ENDPOINT = r'''
@app.post("/analyze-nifti")
async def analyze_nifti(file: UploadFile = File(...)):
    """
    Perform 3D NIfTI cardiac MRI segmentation
    using the CardioSeg AI 2D U-Net slice-by-slice.
    """

    filename = file.filename.lower()

    if not (
        filename.endswith(".nii")
        or filename.endswith(".nii.gz")
    ):
        raise HTTPException(
            status_code=400,
            detail="Please upload a .nii or .nii.gz MRI file."
        )

    save_path = os.path.join(
        UPLOAD_DIR,
        file.filename
    )

    with open(save_path, "wb") as buffer:
        shutil.copyfileobj(
            file.file,
            buffer
        )

    try:

        # Load NIfTI
        mri_nii = nib.load(save_path)

        volume = mri_nii.get_fdata().astype(
            np.float32
        )

        if volume.ndim != 3:
            raise HTTPException(
                status_code=400,
                detail=(
                    f"Expected a 3D MRI volume, "
                    f"but received shape {volume.shape}"
                )
            )

        original_shape = volume.shape

        # Create output mask
        predicted_volume = np.zeros(
            original_shape,
            dtype=np.uint8
        )

        # Process every slice
        for slice_index in range(
            original_shape[2]
        ):

            image_slice = volume[
                :,
                :,
                slice_index
            ]

            # Normalize
            min_value = np.min(
                image_slice
            )

            max_value = np.max(
                image_slice
            )

            if max_value > min_value:

                normalized = (
                    image_slice - min_value
                ) / (
                    max_value - min_value
                )

            else:

                normalized = np.zeros_like(
                    image_slice
                )

            # Resize to model input
            resized = cv2.resize(
                normalized,
                (128, 128),
                interpolation=cv2.INTER_LINEAR
            )

            # Model input
            model_input = np.expand_dims(
                resized,
                axis=(0, -1)
            )

            # Prediction
            probabilities = (
                CARDIOSEG_MODEL.predict(
                    model_input,
                    verbose=0
                )[0]
            )

            predicted_small = np.argmax(
                probabilities,
                axis=-1
            ).astype(np.uint8)

            # Resize prediction back
            # to original MRI dimensions
            predicted_original = cv2.resize(
                predicted_small,
                (
                    original_shape[1],
                    original_shape[0]
                ),
                interpolation=cv2.INTER_NEAREST
            )

            predicted_volume[
                :,
                :,
                slice_index
            ] = predicted_original

        # Save predicted segmentation
        base_name = os.path.basename(
            file.filename
        )

        if base_name.endswith(".nii.gz"):
            base_name = base_name[
                :-7
            ]
        elif base_name.endswith(".nii"):
            base_name = base_name[
                :-4
            ]

        output_filename = (
            base_name +
            "_predicted_mask.nii.gz"
        )

        output_path = os.path.join(
            RESULT_DIR,
            output_filename
        )

        prediction_nii = nib.Nifti1Image(
            predicted_volume,
            mri_nii.affine,
            mri_nii.header
        )

        nib.save(
            prediction_nii,
            output_path
        )

        # Class voxel counts
        background_voxels = int(
            np.sum(
                predicted_volume == 0
            )
        )

        rv_voxels = int(
            np.sum(
                predicted_volume == 1
            )
        )

        myocardium_voxels = int(
            np.sum(
                predicted_volume == 2
            )
        )

        lv_voxels = int(
            np.sum(
                predicted_volume == 3
            )
        )

        print(
            f"NIfTI segmentation completed: "
            f"{file.filename}"
        )

        return {
            "status": "success",
            "filename": file.filename,
            "file_type": "NIfTI",
            "input_shape": list(
                original_shape
            ),
            "voxel_spacing_mm": [
                float(x)
                for x in mri_nii.header.get_zooms()[:3]
            ],
            "model": "CardioSeg AI",
            "architecture": "2D U-Net",
            "classes": {
                "0": "Background",
                "1": "RV",
                "2": "Myocardium",
                "3": "LV"
            },
            "predicted_mask": (
                output_filename
            ),
            "predicted_mask_path": (
                output_path
            ),
            "voxel_counts": {
                "background": background_voxels,
                "RV": rv_voxels,
                "myocardium": myocardium_voxels,
                "LV": lv_voxels
            },
            "message": (
                "3D NIfTI segmentation "
                "completed successfully."
            )
        }

    except HTTPException:
        raise

    except Exception as e:

        if os.path.exists(save_path):
            os.remove(save_path)

        raise HTTPException(
            status_code=500,
            detail=(
                f"NIfTI segmentation failed: "
                f"{str(e)}"
            )
        )
'''

    with open(MAIN_PY, "w") as f:
        f.write(
            content_before_endpoint +
            NEW_ENDPOINT
        )

    print(
        "✅ /analyze-nifti endpoint "
        "upgraded to AI segmentation."
    )

    print(
        "New file size:",
        os.path.getsize(MAIN_PY),
        "bytes"
    )
```

```python
import nibabel as nib
import numpy as np
import cv2

# Select one patient
patient_id = "patient001"
patient_dir = os.path.join(TRAIN_DIR, patient_id)

# Find a labeled frame
files = os.listdir(patient_dir)

gt_files = [
    f for f in files
    if f.endswith("_gt.nii.gz")
]

print("Ground-truth files:", gt_files)

# Use the first labeled frame
gt_file = sorted(gt_files)[0]
image_file = gt_file.replace("_gt.nii.gz", ".nii.gz")

image_path = os.path.join(patient_dir, image_file)

print("MRI file:", image_file)

# Load MRI
image_nii = nib.load(image_path)
image_data = image_nii.get_fdata()

print("Original MRI shape:", image_data.shape)

# Select middle slice
slice_index = image_data.shape[2] // 2

image_slice = image_data[:, :, slice_index].astype(np.float32)

# Normalize
min_value = np.min(image_slice)
max_value = np.max(image_slice)

if max_value > min_value:
    image_normalized = (
        image_slice - min_value
    ) / (max_value - min_value)
else:
    image_normalized = np.zeros_like(image_slice)

# Resize to model input
image_resized = cv2.resize(
    image_normalized,
    (128, 128),
    interpolation=cv2.INTER_LINEAR
)

# Add batch + channel dimensions
image_input = np.expand_dims(
    image_resized,
    axis=(0, -1)
).astype(np.float32)

print("Model input shape:", image_input.shape)

# Prediction
prediction = model.predict(
    image_input,
    verbose=0
)

print("Prediction shape:", prediction.shape)

# Convert probabilities to class labels
predicted_mask = np.argmax(
    prediction[0],
    axis=-1
).astype(np.uint8)

print("Predicted mask shape:", predicted_mask.shape)
print("Predicted labels:", np.unique(predicted_mask))
```

```python
import matplotlib.pyplot as plt

# Ground-truth mask for comparison
mask_file = gt_file
mask_path = os.path.join(patient_dir, mask_file)

mask_nii = nib.load(mask_path)
mask_data = mask_nii.get_fdata()

mask_slice = mask_data[:, :, slice_index].astype(np.uint8)

# Resize ground truth using nearest-neighbor
mask_resized = cv2.resize(
    mask_slice,
    (128, 128),
    interpolation=cv2.INTER_NEAREST
)

# Create figure
plt.figure(figsize=(16, 4))

plt.subplot(1, 4, 1)
plt.imshow(image_resized, cmap="gray")
plt.title("MRI")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(mask_resized, cmap="viridis")
plt.title("Ground Truth")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(predicted_mask, cmap="viridis")
plt.title("Prediction")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(image_resized, cmap="gray")
plt.imshow(predicted_mask, cmap="viridis", alpha=0.45)
plt.title("Prediction Overlay")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Step 6: Calculate class-wise Dice score

def calculate_dice(y_true, y_pred, class_id, smooth=1e-6):
    true_mask = (y_true == class_id)
    pred_mask = (y_pred == class_id)

    intersection = np.sum(
        true_mask & pred_mask
    )

    denominator = (
        np.sum(true_mask) +
        np.sum(pred_mask)
    )

    dice = (
        2.0 * intersection + smooth
    ) / (
        denominator + smooth
    )

    return dice


class_names = {
    0: "Background",
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

print("Class-wise Dice Scores")
print("-" * 35)

dice_scores = {}

for class_id, class_name in class_names.items():

    dice = calculate_dice(
        mask_resized,
        predicted_mask,
        class_id
    )

    dice_scores[class_name] = dice

    print(
        f"{class_name:15s}: "
        f"{dice * 100:.2f}%"
    )

# Overall Dice
overall_dice = np.mean(
    list(dice_scores.values())
)

print("-" * 35)
print(
    f"Mean class Dice: "
    f"{overall_dice * 100:.2f}%"
)
```

```python
# Step 7: Calculate anatomical area

# Pixel spacing from the original ACDC MRI
pixel_spacing = image_nii.header.get_zooms()[:2]

spacing_x = float(pixel_spacing[0])
spacing_y = float(pixel_spacing[1])

pixel_area_mm2 = spacing_x * spacing_y

print("Pixel spacing:")
print(f"X: {spacing_x:.4f} mm")
print(f"Y: {spacing_y:.4f} mm")
print(f"Pixel area: {pixel_area_mm2:.4f} mm²")
print()

area_results = {}

for class_id, class_name in {
    1: "RV",
    2: "Myocardium",
    3: "LV"
}.items():

    pixel_count = np.sum(predicted_mask == class_id)

    area_mm2 = pixel_count * pixel_area_mm2
    area_cm2 = area_mm2 / 100.0

    area_results[class_name] = area_cm2

    print(f"{class_name}:")
    print(f"  Pixels   : {pixel_count}")
    print(f"  Area     : {area_mm2:.2f} mm²")
    print(f"  Area     : {area_cm2:.2f} cm²")
    print()
```

```python
# Step 8: Calculate 3D cardiac volume

# Get voxel spacing from the original MRI
voxel_spacing = image_nii.header.get_zooms()

pixel_spacing_x = float(voxel_spacing[0])
pixel_spacing_y = float(voxel_spacing[1])
slice_thickness = float(voxel_spacing[2])

pixel_area_mm2 = pixel_spacing_x * pixel_spacing_y

print("Voxel spacing:")
print(f"X spacing       : {pixel_spacing_x:.4f} mm")
print(f"Y spacing       : {pixel_spacing_y:.4f} mm")
print(f"Slice thickness : {slice_thickness:.4f} mm")
print()

# Arrays to store areas for every slice
rv_areas = []
myo_areas = []
lv_areas = []

num_slices = image_data.shape[2]

for z in range(num_slices):

    # Get MRI slice
    current_slice = image_data[:, :, z].astype(np.float32)

    # Normalize
    min_value = np.min(current_slice)
    max_value = np.max(current_slice)

    if max_value > min_value:
        current_slice = (
            current_slice - min_value
        ) / (max_value - min_value)
    else:
        current_slice = np.zeros_like(current_slice)

    # Resize
    current_slice = cv2.resize(
        current_slice,
        (128, 128),
        interpolation=cv2.INTER_LINEAR
    )

    # Model input
    current_input = np.expand_dims(
        current_slice,
        axis=(0, -1)
    ).astype(np.float32)

    # Prediction
    current_prediction = model.predict(
        current_input,
        verbose=0
    )

    current_mask = np.argmax(
        current_prediction[0],
        axis=-1
    ).astype(np.uint8)

    # Count pixels
    rv_pixels = np.sum(current_mask == 1)
    myo_pixels = np.sum(current_mask == 2)
    lv_pixels = np.sum(current_mask == 3)

    # Convert pixels to area
    rv_area = rv_pixels * pixel_area_mm2
    myo_area = myo_pixels * pixel_area_mm2
    lv_area = lv_pixels * pixel_area_mm2

    rv_areas.append(rv_area)
    myo_areas.append(myo_area)
    lv_areas.append(lv_area)


# Calculate volumes
rv_volume_mm3 = np.sum(rv_areas) * slice_thickness
myo_volume_mm3 = np.sum(myo_areas) * slice_thickness
lv_volume_mm3 = np.sum(lv_areas) * slice_thickness

# Convert mm³ → mL
rv_volume_ml = rv_volume_mm3 / 1000
myo_volume_ml = myo_volume_mm3 / 1000
lv_volume_ml = lv_volume_mm3 / 1000

print("Predicted Cardiac Volumes")
print("-" * 40)

print(f"RV volume         : {rv_volume_ml:.2f} mL")
print(f"Myocardium volume : {myo_volume_ml:.2f} mL")
print(f"LV volume         : {lv_volume_ml:.2f} mL")
```

```python
# Step 9: Calculate EDV, ESV and LVEF

def calculate_lv_volume(image_path):

    nii = nib.load(image_path)
    data = nii.get_fdata().astype(np.float32)

    spacing = nii.header.get_zooms()

    pixel_area_mm2 = float(spacing[0]) * float(spacing[1])
    slice_thickness = float(spacing[2])

    total_lv_area = 0.0

    for z in range(data.shape[2]):

        current_slice = data[:, :, z]

        # Normalize
        min_value = np.min(current_slice)
        max_value = np.max(current_slice)

        if max_value > min_value:
            current_slice = (
                current_slice - min_value
            ) / (max_value - min_value)
        else:
            current_slice = np.zeros_like(current_slice)

        # Resize
        current_slice = cv2.resize(
            current_slice,
            (128, 128),
            interpolation=cv2.INTER_LINEAR
        )

        # Model input
        model_input = np.expand_dims(
            current_slice,
            axis=(0, -1)
        ).astype(np.float32)

        # Prediction
        prediction = model.predict(
            model_input,
            verbose=0
        )

        predicted_mask = np.argmax(
            prediction[0],
            axis=-1
        ).astype(np.uint8)

        # LV = class 3
        lv_pixels = np.sum(predicted_mask == 3)

        lv_area_mm2 = lv_pixels * pixel_area_mm2

        total_lv_area += lv_area_mm2

    # Area × slice thickness = volume
    volume_mm3 = total_lv_area * slice_thickness

    # 1000 mm³ = 1 mL
    volume_ml = volume_mm3 / 1000.0

    return volume_ml


# Find cardiac frames
frame_files = [
    f for f in os.listdir(patient_dir)
    if f.endswith(".nii.gz") and "_gt" not in f
]

print("Available MRI frames:")
for f in sorted(frame_files):
    print(f)

print()

# Calculate volume for each frame
frame_volumes = {}

for frame_file in sorted(frame_files):

    image_path = os.path.join(
        patient_dir,
        frame_file
    )

    volume = calculate_lv_volume(image_path)

    frame_volumes[frame_file] = volume

    print(
        f"{frame_file}: "
        f"LV Volume = {volume:.2f} mL"
    )


# Determine EDV and ESV
edv_frame = max(
    frame_volumes,
    key=frame_volumes.get
)

esv_frame = min(
    frame_volumes,
    key=frame_volumes.get
)

EDV = frame_volumes[edv_frame]
ESV = frame_volumes[esv_frame]

# Calculate LVEF
LVEF = ((EDV - ESV) / EDV) * 100

print()
print("=" * 45)
print("CARDIAC FUNCTION RESULTS")
print("=" * 45)

print(f"EDV  : {EDV:.2f} mL")
print(f"ESV  : {ESV:.2f} mL")
print(f"LVEF : {LVEF:.2f}%")
print()
print(f"EDV frame: {edv_frame}")
print(f"ESV frame: {esv_frame}")
```

```python
# Step 9: Calculate EDV, ESV and LVEF
# Only process labeled cardiac frames, not the 4D file.

def calculate_lv_volume(image_path):

    nii = nib.load(image_path)
    data = nii.get_fdata().astype(np.float32)

    spacing = nii.header.get_zooms()

    pixel_area_mm2 = float(spacing[0]) * float(spacing[1])
    slice_thickness = float(spacing[2])

    total_lv_area_mm2 = 0.0

    for z in range(data.shape[2]):

        current_slice = data[:, :, z]

        # Normalize
        min_value = np.min(current_slice)
        max_value = np.max(current_slice)

        if max_value > min_value:
            current_slice = (
                current_slice - min_value
            ) / (max_value - min_value)
        else:
            current_slice = np.zeros_like(current_slice)

        # Resize to model input size
        current_slice = cv2.resize(
            current_slice,
            (128, 128),
            interpolation=cv2.INTER_LINEAR
        )

        # Explicit model input shape:
        # (batch, height, width, channel)
        model_input = current_slice[np.newaxis, :, :, np.newaxis]
        model_input = model_input.astype(np.float32)

        # Prediction
        prediction = model.predict(
            model_input,
            verbose=0
        )

        # Convert probabilities to class labels
        current_mask = np.argmax(
            prediction[0],
            axis=-1
        ).astype(np.uint8)

        # LV = class 3
        lv_pixels = np.sum(current_mask == 3)

        # Pixel count → area
        lv_area_mm2 = lv_pixels * pixel_area_mm2

        total_lv_area_mm2 += lv_area_mm2

    # Area × slice thickness
    volume_mm3 = total_lv_area_mm2 * slice_thickness

    # mm³ → mL
    volume_ml = volume_mm3 / 1000.0

    return volume_ml


# ------------------------------------------------
# Only use the two labeled cardiac frames
# ------------------------------------------------

frame_files = [
    "patient001_frame01.nii.gz",
    "patient001_frame12.nii.gz"
]

frame_volumes = {}

print("Calculating LV volumes...")
print()

for frame_file in frame_files:

    image_path = os.path.join(
        patient_dir,
        frame_file
    )

    if not os.path.exists(image_path):
        print(f"File not found: {frame_file}")
        continue

    volume = calculate_lv_volume(image_path)

    frame_volumes[frame_file] = volume

    print(
        f"{frame_file} → "
        f"LV Volume = {volume:.2f} mL"
    )


# ------------------------------------------------
# Calculate EDV and ESV
# ------------------------------------------------

if len(frame_volumes) == 2:

    edv_frame = max(
        frame_volumes,
        key=frame_volumes.get
    )

    esv_frame = min(
        frame_volumes,
        key=frame_volumes.get
    )

    EDV = frame_volumes[edv_frame]
    ESV = frame_volumes[esv_frame]

    if EDV > 0:
        LVEF = ((EDV - ESV) / EDV) * 100
    else:
        LVEF = 0

    print()
    print("=" * 45)
    print("CARDIAC FUNCTION RESULTS")
    print("=" * 45)

    print(f"EDV  : {EDV:.2f} mL")
    print(f"ESV  : {ESV:.2f} mL")
    print(f"LVEF : {LVEF:.2f}%")

    print()
    print(f"EDV frame: {edv_frame}")
    print(f"ESV frame: {esv_frame}")

else:
    print()
    print("Could not calculate EDV/ESV.")
    print("Both frame01 and frame12 are required.")
```

```python
# Step 10A: Correct probability maps for the selected MRI slice

# Recreate the model input for the selected slice
image_input = np.expand_dims(
    image_resized,
    axis=(0, -1)
).astype(np.float32)

# Predict specifically for this slice
slice_prediction = model.predict(
    image_input,
    verbose=0
)

# Extract probabilities
slice_probabilities = slice_prediction[0]

# Final segmentation
slice_predicted_mask = np.argmax(
    slice_probabilities,
    axis=-1
).astype(np.uint8)

# Visualize
plt.figure(figsize=(16, 4))

plt.subplot(1, 5, 1)
plt.imshow(image_resized, cmap="gray")
plt.title("MRI")
plt.axis("off")

plt.subplot(1, 5, 2)
plt.imshow(
    slice_probabilities[:, :, 1],
    cmap="hot",
    vmin=0,
    vmax=1
)
plt.title("RV Probability")
plt.axis("off")

plt.subplot(1, 5, 3)
plt.imshow(
    slice_probabilities[:, :, 2],
    cmap="hot",
    vmin=0,
    vmax=1
)
plt.title("Myocardium Probability")
plt.axis("off")

plt.subplot(1, 5, 4)
plt.imshow(
    slice_probabilities[:, :, 3],
    cmap="hot",
    vmin=0,
    vmax=1
)
plt.title("LV Probability")
plt.axis("off")

plt.subplot(1, 5, 5)
plt.imshow(
    slice_predicted_mask,
    cmap="viridis"
)
plt.title("Final Segmentation")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Step 11: Generate uncertainty map using entropy

# Get probabilities for all 4 classes
probabilities = slice_probabilities

# Small value to avoid log(0)
epsilon = 1e-8

# Entropy calculation
uncertainty_map = -np.sum(
    probabilities * np.log(probabilities + epsilon),
    axis=-1
)

# Normalize uncertainty to 0-1
max_entropy = np.log(4)

uncertainty_normalized = (
    uncertainty_map / max_entropy
)

# Display
plt.figure(figsize=(16, 4))

plt.subplot(1, 4, 1)
plt.imshow(
    image_resized,
    cmap="gray"
)
plt.title("MRI")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(
    slice_predicted_mask,
    cmap="viridis"
)
plt.title("Segmentation")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(
    uncertainty_normalized,
    cmap="hot",
    vmin=0,
    vmax=1
)
plt.colorbar(
    fraction=0.046,
    pad=0.04
)
plt.title("Uncertainty Map")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(
    image_resized,
    cmap="gray"
)
plt.imshow(
    uncertainty_normalized,
    cmap="hot",
    alpha=0.55,
    vmin=0,
    vmax=1
)
plt.title("Uncertainty Overlay")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Uncertainty map generated successfully.")
print(
    f"Mean uncertainty: "
    f"{np.mean(uncertainty_normalized):.4f}"
)
print(
    f"Maximum uncertainty: "
    f"{np.max(uncertainty_normalized):.4f}"
)
```

```python
# Step 12: Seg-Grad-CAM for LV segmentation

# Find the last convolutional layer
conv_layers = [
    layer for layer in model.layers
    if isinstance(layer, tf.keras.layers.Conv2D)
]

last_conv_layer = conv_layers[-1]

print("Grad-CAM layer:", last_conv_layer.name)
print("Layer output shape:", last_conv_layer.output.shape)


# Create a model that returns:
# 1. Last convolutional feature maps
# 2. Final segmentation output

grad_model = tf.keras.models.Model(
    inputs=model.inputs,
    outputs=[
        last_conv_layer.output,
        model.output
    ]
)


# Calculate gradients
with tf.GradientTape() as tape:

    conv_outputs, predictions = grad_model(
        image_input,
        training=False
    )

    # LV = class 3
    lv_probability = predictions[:, :, :, 3]

    # Sum LV probabilities
    target = tf.reduce_sum(lv_probability)

# Gradient of LV prediction
grads = tape.gradient(
    target,
    conv_outputs
)

# Global average pooling of gradients
pooled_grads = tf.reduce_mean(
    grads,
    axis=(1, 2)
)

# Remove batch dimension
conv_outputs = conv_outputs[0]
pooled_grads = pooled_grads[0]

# Weight feature maps using gradients
gradcam = tf.reduce_sum(
    conv_outputs * pooled_grads,
    axis=-1
)

# ReLU
gradcam = tf.maximum(
    gradcam,
    0
)

# Normalize
max_value = tf.reduce_max(gradcam)

if max_value > 0:
    gradcam = gradcam / max_value

gradcam = gradcam.numpy()

# Resize Grad-CAM to MRI size
gradcam_resized = cv2.resize(
    gradcam,
    (128, 128),
    interpolation=cv2.INTER_LINEAR
)


# ------------------------------------------------
# Visualization
# ------------------------------------------------

plt.figure(figsize=(16, 4))

plt.subplot(1, 4, 1)
plt.imshow(
    image_resized,
    cmap="gray"
)
plt.title("MRI")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(
    slice_predicted_mask,
    cmap="viridis"
)
plt.title("LV Segmentation")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(
    gradcam_resized,
    cmap="jet",
    vmin=0,
    vmax=1
)
plt.colorbar(
    fraction=0.046,
    pad=0.04
)
plt.title("Seg-Grad-CAM")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(
    image_resized,
    cmap="gray"
)
plt.imshow(
    gradcam_resized,
    cmap="jet",
    alpha=0.5,
    vmin=0,
    vmax=1
)
plt.title("Grad-CAM Overlay")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Seg-Grad-CAM generated successfully.")
print("Target class: LV (class 3)")
```

```python
# Step 12C: Correct Seg-Grad-CAM using conv2d_35

# Use the last internal convolutional layer
last_conv_layer = model.get_layer("conv2d_35")

print("Grad-CAM layer:", last_conv_layer.name)
print("Feature map shape:", last_conv_layer.output.shape)

# Create Grad-CAM model
grad_model = tf.keras.models.Model(
    inputs=model.input,
    outputs=[
        last_conv_layer.output,
        model.output
    ]
)

# Make sure input has the expected shape
gradcam_input = tf.convert_to_tensor(
    image_input,
    dtype=tf.float32
)

# Calculate gradients
with tf.GradientTape() as tape:

    conv_outputs, predictions = grad_model(
        {"mri_input": gradcam_input},
        training=False
    )

    # LV = class 3
    lv_probability = predictions[:, :, :, 3]

    # Focus Grad-CAM on LV prediction
    target = tf.reduce_sum(lv_probability)

# Gradient of LV prediction
grads = tape.gradient(
    target,
    conv_outputs
)

# Global average pooling
pooled_grads = tf.reduce_mean(
    grads,
    axis=(1, 2)
)

# Remove batch dimension
conv_outputs = conv_outputs[0]
pooled_grads = pooled_grads[0]

# Weight feature maps
gradcam = tf.reduce_sum(
    conv_outputs * pooled_grads,
    axis=-1
)

# Keep positive activations
gradcam = tf.maximum(
    gradcam,
    0
)

# Normalize
max_value = tf.reduce_max(gradcam)

gradcam = tf.where(
    max_value > 0,
    gradcam / max_value,
    gradcam
)

gradcam = gradcam.numpy()

# Resize to MRI dimensions
gradcam_resized = cv2.resize(
    gradcam,
    (128, 128),
    interpolation=cv2.INTER_LINEAR
)

# ------------------------------------------------
# Visualization
# ------------------------------------------------

plt.figure(figsize=(16, 4))

plt.subplot(1, 4, 1)

plt.imshow(
    image_resized,
    cmap="gray"
)

plt.title("MRI")
plt.axis("off")


plt.subplot(1, 4, 2)

plt.imshow(
    slice_predicted_mask,
    cmap="viridis"
)

plt.title("LV Segmentation")
plt.axis("off")


plt.subplot(1, 4, 3)

plt.imshow(
    gradcam_resized,
    cmap="jet",
    vmin=0,
    vmax=1
)

plt.colorbar(
    fraction=0.046,
    pad=0.04
)

plt.title("Seg-Grad-CAM")
plt.axis("off")


plt.subplot(1, 4, 4)

plt.imshow(
    image_resized,
    cmap="gray"
)

plt.imshow(
    gradcam_resized,
    cmap="jet",
    alpha=0.5,
    vmin=0,
    vmax=1
)

plt.title("Grad-CAM Overlay")
plt.axis("off")

plt.tight_layout()
plt.show()

print()
print("Seg-Grad-CAM generated successfully.")
print("Target class: LV (class 3)")
print(f"Maximum activation: {np.max(gradcam_resized):.4f}")
print(f"Mean activation: {np.mean(gradcam_resized):.4f}")
```

```python
# Step 15: Create FastAPI application

app_code = r'''
from fastapi import FastAPI, UploadFile, File
from fastapi.middleware.cors import CORSMiddleware
import tensorflow as tf
import numpy as np
import cv2
import os
import tempfile
import nibabel as nib

# --------------------------------------------------
# Paths
# --------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/CardioSeg_AI"

MODEL_PATH = os.path.join(
    BASE_DIR,
    "models",
    "CardioSeg_AI_Boundary_best.keras"
)

# --------------------------------------------------
# Load model
# --------------------------------------------------

print("Loading CardioSeg AI model...")

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

print("Model loaded successfully.")
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)


# --------------------------------------------------
# FastAPI application
# --------------------------------------------------

app = FastAPI(
    title="CardioSeg AI API",
    description="AI-based cardiac MRI segmentation API",
    version="1.0.0"
)


# --------------------------------------------------
# CORS
# --------------------------------------------------

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)


# --------------------------------------------------
# Root endpoint
# --------------------------------------------------

@app.get("/")
def root():

    return {
        "project": "CardioSeg AI",
        "status": "running",
        "model": "Boundary-aware U-Net",
        "input_size": "128x128",
        "classes": {
            "0": "Background",
            "1": "RV",
            "2": "Myocardium",
            "3": "LV"
        }
    }


# --------------------------------------------------
# Health endpoint
# --------------------------------------------------

@app.get("/health")
def health():

    return {
        "status": "healthy",
        "model_loaded": model is not None
    }


# --------------------------------------------------
# Model information
# --------------------------------------------------

@app.get("/model-info")
def model_info():

    return {
        "model_name": "CardioSeg AI Boundary-aware U-Net",
        "input_shape": str(model.input_shape),
        "output_shape": str(model.output_shape),
        "num_classes": 4,
        "classes": [
            "Background",
            "RV",
            "Myocardium",
            "LV"
        ]
    }


# --------------------------------------------------
# MRI prediction endpoint
# --------------------------------------------------

@app.post("/predict")
async def predict(file: UploadFile = File(...)):

    # Create temporary file
    suffix = os.path.splitext(file.filename)[1]

    with tempfile.NamedTemporaryFile(
        delete=False,
        suffix=suffix
    ) as temp_file:

        contents = await file.read()
        temp_file.write(contents)
        temp_path = temp_file.name

    try:

        # --------------------------------------------------
        # Load NIfTI MRI
        # --------------------------------------------------

        nii = nib.load(temp_path)

        image_data = nii.get_fdata().astype(
            np.float32
        )

        # If 4D, use first time frame
        if image_data.ndim == 4:
            image_data = image_data[:, :, :, 0]

        num_slices = image_data.shape[2]

        predictions = []

        # --------------------------------------------------
        # Process every slice
        # --------------------------------------------------

        for z in range(num_slices):

            image_slice = image_data[:, :, z]

            # Normalize
            min_value = np.min(image_slice)
            max_value = np.max(image_slice)

            if max_value > min_value:

                image_normalized = (
                    image_slice - min_value
                ) / (
                    max_value - min_value
                )

            else:

                image_normalized = np.zeros_like(
                    image_slice
                )

            # Resize
            image_resized = cv2.resize(
                image_normalized,
                (128, 128),
                interpolation=cv2.INTER_LINEAR
            )

            # Model input
            image_input = np.expand_dims(
                image_resized,
                axis=(0, -1)
            ).astype(np.float32)

            # Prediction
            prediction = model.predict(
                image_input,
                verbose=0
            )

            # Class mask
            predicted_mask = np.argmax(
                prediction[0],
                axis=-1
            )

            predictions.append(
                predicted_mask.tolist()
            )

        # --------------------------------------------------
        # Response
        # --------------------------------------------------

        return {
            "status": "success",
            "filename": file.filename,
            "original_shape": list(image_data.shape),
            "number_of_slices": num_slices,
            "prediction_shape": [
                num_slices,
                128,
                128
            ],
            "classes": {
                "0": "Background",
                "1": "RV",
                "2": "Myocardium",
                "3": "LV"
            },
            "predictions": predictions
        }

    finally:

        if os.path.exists(temp_path):
            os.remove(temp_path)
'''

app_path = os.path.join(
    BACKEND_DIR,
    "app.py"
)

with open(app_path, "w") as f:
    f.write(app_code)

print("FastAPI application created successfully!")
print()
print("File:", app_path)
print("Exists:", os.path.exists(app_path))
```

```python
app_path = os.path.join(
    BASE_DIR,
    "backend",
    "app.py"
)

with open(app_path, "r") as f:
    code = f.read()

old_code = '''suffix = os.path.splitext(file.filename)[1]
    with tempfile.NamedTemporaryFile(delete=False, suffix=suffix) as temp_file:
'''

new_code = '''suffix = ".nii.gz" if file.filename.endswith(".nii.gz") else os.path.splitext(file.filename)[1]
    with tempfile.NamedTemporaryFile(delete=False, suffix=suffix) as temp_file:
'''

if old_code in code:
    code = code.replace(old_code, new_code)

    with open(app_path, "w") as f:
        f.write(code)

    print("Backend file handling fixed successfully.")
else:
    print("The expected code section was not found.")
```

```python
# Fix the /predict endpoint for .nii.gz files

app_path = os.path.join(
    BASE_DIR,
    "backend",
    "app.py"
)

with open(app_path, "r") as f:
    code = f.read()

start = code.find("@app.post(\"/predict\")")

if start == -1:
    print("Could not find the /predict endpoint.")
else:
    new_predict = '''@app.post("/predict")
async def predict(file: UploadFile = File(...)):
    # Correctly preserve .nii.gz extension
    if file.filename.endswith(".nii.gz"):
        suffix = ".nii.gz"
    else:
        suffix = os.path.splitext(file.filename)[1]

    with tempfile.NamedTemporaryFile(
        delete=False,
        suffix=suffix
    ) as temp_file:

        contents = await file.read()
        temp_file.write(contents)
        temp_path = temp_file.name

    try:
        # Load NIfTI file
        nii = nib.load(temp_path)
        image_data = nii.get_fdata().astype(np.float32)

        # Handle possible 4D input
        if image_data.ndim == 4:
            image_data = image_data[:, :, :, 0]

        num_slices = image_data.shape[2]
        predictions = []

        for z in range(num_slices):

            image_slice = image_data[:, :, z]

            min_value = np.min(image_slice)
            max_value = np.max(image_slice)

            if max_value > min_value:
                image_normalized = (
                    image_slice - min_value
                ) / (
                    max_value - min_value
                )
            else:
                image_normalized = np.zeros_like(
                    image_slice
                )

            image_resized = cv2.resize(
                image_normalized,
                (128, 128),
                interpolation=cv2.INTER_LINEAR
            )

            image_input = np.expand_dims(
                image_resized,
                axis=(0, -1)
            ).astype(np.float32)

            prediction = model.predict(
                image_input,
                verbose=0
            )

            predicted_mask = np.argmax(
                prediction[0],
                axis=-1
            ).astype(np.uint8)

            predictions.append(
                predicted_mask.tolist()
            )

        return {
            "status": "success",
            "filename": file.filename,
            "original_shape": list(image_data.shape),
            "number_of_slices": num_slices,
            "prediction_shape": [
                num_slices,
                128,
                128
            ],
            "classes": {
                "0": "Background",
                "1": "RV",
                "2": "Myocardium",
                "3": "LV"
            },
            "predictions": predictions
        }

    finally:
        if os.path.exists(temp_path):
            os.remove(temp_path)
'''

    code = code[:start] + new_predict + "\n"

    with open(app_path, "w") as f:
        f.write(code)

    print("Successfully replaced the /predict endpoint.")
    print("Correct .nii.gz handling has been added.")
```

```python
%cd /content/cardioseg_frontend

from pathlib import Path

app_file = Path("src/App.tsx")
app_file.write_text(r'''
import { useState } from "react";
import axios from "axios";
import {
  Activity,
  Brain,
  ChevronRight,
  CircleGauge,
  FileImage,
  FileText,
  HeartPulse,
  Layers3,
  ScanHeart,
  ShieldCheck,
  Sparkles,
  Upload,
  Zap,
} from "lucide-react";
import "./App.css";

function App() {
  const [selectedFile, setSelectedFile] = useState<File | null>(null);
  const [uploading, setUploading] = useState(false);
  const [predictionResult, setPredictionResult] = useState<any>(null);
  const [error, setError] = useState("");

  const handleFileChange = (event: React.ChangeEvent<HTMLInputElement>) => {
    const file = event.target.files?.[0];

    if (!file) return;

    setSelectedFile(file);
    setPredictionResult(null);
    setError("");
  };

  const handlePredict = async () => {
    if (!selectedFile) {
      setError("Please select an MRI file first.");
      return;
    }

    setUploading(true);
    setError("");
    setPredictionResult(null);

    try {
      const formData = new FormData();
      formData.append("file", selectedFile);

      const response = await axios.post(
        "http://localhost:8001/predict",
        formData,
        {
          headers: {
            "Content-Type": "multipart/form-data",
          },
        }
      );

      setPredictionResult(response.data);
    } catch (err: any) {
      console.error(err);
      setError(
        err?.response?.data?.detail ||
        "Unable to connect to the CardioSeg AI backend."
      );
    } finally {
      setUploading(false);
    }
  };

  return (
    <div className="app-shell">
      <div className="ambient ambient-one"></div>
      <div className="ambient ambient-two"></div>
      <div className="ambient ambient-three"></div>

      <aside className="sidebar glass-panel">
        <div className="brand">
          <div className="brand-icon">
            <HeartPulse size={24} />
          </div>
          <div>
            <h2>CardioSeg</h2>
            <span>AI</span>
          </div>
        </div>

        <nav className="sidebar-nav">
          <button className="nav-item active">
            <CircleGauge size={19} />
            Dashboard
          </button>

          <button className="nav-item">
            <ScanHeart size={19} />
            Segmentation
          </button>

          <button className="nav-item">
            <Activity size={19} />
            Analysis
          </button>

          <button className="nav-item">
            <FileText size={19} />
            Reports
          </button>
        </nav>

        <div className="sidebar-bottom">
          <div className="model-status">
            <span className="status-dot"></span>
            <div>
              <strong>AI Model Online</strong>
              <small>Boundary-aware U-Net</small>
            </div>
          </div>
        </div>
      </aside>

      <main className="main-content">
        <header className="topbar">
          <div>
            <p className="eyebrow">CARDIAC MRI ANALYSIS</p>
            <h1>CardioSeg AI</h1>
          </div>

          <div className="topbar-status">
            <ShieldCheck size={18} />
            Secure AI Pipeline
          </div>
        </header>

        <section className="hero glass-panel">
          <div className="hero-content">
            <div className="hero-badge">
              <Sparkles size={15} />
              AI-POWERED CARDIAC SEGMENTATION
            </div>

            <h2>
              Understand the heart.
              <br />
              <span>Segment with intelligence.</span>
            </h2>

            <p>
              Analyze cardiac MRI scans with automated segmentation of the
              right ventricle, myocardium, and left ventricle.
            </p>
          </div>

          <div className="heart-visual">
            <div className="orbit orbit-one"></div>
            <div className="orbit orbit-two"></div>
            <div className="orbit orbit-three"></div>

            <div className="heart-core">
              <HeartPulse size={72} strokeWidth={1.4} />
            </div>

            <div className="floating-label label-rv">RV</div>
            <div className="floating-label label-myo">MYO</div>
            <div className="floating-label label-lv">LV</div>
          </div>
        </section>

        <section className="upload-section">
          <div className="section-heading">
            <div>
              <p className="eyebrow">INPUT</p>
              <h3>Upload Cardiac MRI</h3>
            </div>

            <span className="supported-format">
              NIfTI • .nii / .nii.gz
            </span>
          </div>

          <div className="upload-zone glass-panel">
            <input
              id="mri-upload"
              type="file"
              accept=".nii,.nii.gz"
              onChange={handleFileChange}
              hidden
            />

            <label htmlFor="mri-upload" className="upload-content">
              <div className="upload-icon">
                <Upload size={28} />
              </div>

              <h4>
                {selectedFile
                  ? selectedFile.name
                  : "Drop your cardiac MRI here"}
              </h4>

              <p>
                {selectedFile
                  ? `${(selectedFile.size / 1024 / 1024).toFixed(2)} MB selected`
                  : "or click to browse your files"}
              </p>
            </label>

            {selectedFile && (
              <button
                className="primary-button"
                onClick={handlePredict}
                disabled={uploading}
              >
                {uploading ? (
                  <>
                    <Zap size={18} />
                    Processing MRI...
                  </>
                ) : (
                  <>
                    <Brain size={18} />
                    Run AI Segmentation
                    <ChevronRight size={18} />
                  </>
                )}
              </button>
            )}

            {error && <div className="error-message">{error}</div>}
          </div>
        </section>

        <section className="metrics-grid">
          <div className="metric-card glass-panel">
            <div className="metric-icon">
              <Layers3 size={21} />
            </div>
            <span>MODEL INPUT</span>
            <strong>128 × 128</strong>
            <small>2D MRI slices</small>
          </div>

          <div className="metric-card glass-panel">
            <div className="metric-icon">
              <Brain size={21} />
            </div>
            <span>CLASSES</span>
            <strong>4</strong>
            <small>Background + 3 cardiac structures</small>
          </div>

          <div className="metric-card glass-panel">
            <div className="metric-icon">
              <ScanHeart size={21} />
            </div>
            <span>SEGMENTATION</span>
            <strong>U-Net</strong>
            <small>Boundary-aware architecture</small>
          </div>

          <div className="metric-card glass-panel">
            <div className="metric-icon">
              <ShieldCheck size={21} />
            </div>
            <span>PIPELINE</span>
            <strong>AI Ready</strong>
            <small>FastAPI inference backend</small>
          </div>
        </section>

        {predictionResult && (
          <section className="results-panel glass-panel">
            <div className="section-heading">
              <div>
                <p className="eyebrow">RESULT</p>
                <h3>Segmentation Completed</h3>
              </div>
            </div>

            <div className="result-grid">
              <div>
                <span>File</span>
                <strong>{predictionResult.filename}</strong>
              </div>

              <div>
                <span>Original Shape</span>
                <strong>
                  {predictionResult.original_shape?.join(" × ")}
                </strong>
              </div>

              <div>
                <span>Slices</span>
                <strong>{predictionResult.number_of_slices}</strong>
              </div>

              <div>
                <span>Prediction Shape</span>
                <strong>
                  {predictionResult.prediction_shape?.join(" × ")}
                </strong>
              </div>
            </div>
          </section>
        )}

        <section className="analysis-preview">
          <div className="section-heading">
            <div>
              <p className="eyebrow">ANALYSIS</p>
              <h3>Clinical Metrics</h3>
            </div>
          </div>

          <div className="clinical-grid">
            <div className="clinical-card glass-panel">
              <span>EDV</span>
              <strong>—</strong>
              <small>mL</small>
            </div>

            <div className="clinical-card glass-panel">
              <span>ESV</span>
              <strong>—</strong>
              <small>mL</small>
            </div>

            <div className="clinical-card glass-panel">
              <span>LVEF</span>
              <strong>—</strong>
              <small>%</small>
            </div>

            <div className="clinical-card glass-panel">
              <span>Dice Score</span>
              <strong>—</strong>
              <small>Segmentation quality</small>
            </div>
          </div>
        </section>

        <footer className="footer">
          <FileImage size={16} />
          CardioSeg AI • Automated Cardiac MRI Segmentation
        </footer>
      </main>
    </div>
  );
}

export default App;
''')

print("App.tsx updated successfully!")
```

```python
import requests
import os

MRI_PATH = "/content/drive/.shortcut-targets-by-id/1xHjQpDRUHNLEFghhG5U2RBmLRRbxffH7/CardioSeg_AI/dataset/ACDC/database/training/patient001/patient001_frame01.nii.gz"

print("MRI exists:", os.path.exists(MRI_PATH))

with open(MRI_PATH, "rb") as f:
    response = requests.post(
        "http://localhost:8001/predict",
        files={
            "file": (
                "patient001_frame01.nii.gz",
                f,
                "application/gzip"
            )
        }
    )

print("Status:", response.status_code)

if response.status_code == 200:
    result = response.json()

    print("Prediction successful! ✅")
    print("Filename:", result["filename"])
    print("Original shape:", result["original_shape"])
    print("Number of slices:", result["number_of_slices"])
    print("Prediction shape:", result["prediction_shape"])
else:
    print("Error:")
    print(response.text)
```

```python
from pathlib import Path

APP_PATH = Path(
    "/content/drive/.shortcut-targets-by-id/1xHjQpDRUHNLEFghhG5U2RBmLRRbxffH7/"
    "CardioSeg_AI/backend/app.py"
)

content = APP_PATH.read_text()

# Add required imports if they are not already present
if "from fastapi.responses import FileResponse" not in content:
    content = content.replace(
        "from fastapi import",
        "from fastapi.responses import FileResponse\nfrom fastapi import",
        1
    )

if "from fastapi.staticfiles import StaticFiles" not in content:
    content = content.replace(
        "from fastapi import",
        "from fastapi.staticfiles import StaticFiles\nfrom fastapi import",
        1
    )

# Replace the existing root endpoint
old_root = '''@app.get("/")
def root():

    return {
        "project": "CardioSeg AI",
        "status": "running",
        "model": "Boundary-aware U-Net",
        "input_size": "128x128",
        "classes": {
            "0": "Background",
            "1": "RV",
            "2": "Myocardium",
            "3": "LV"
        }
    }
'''

new_root = '''@app.get("/")
def root():
    return FileResponse(
        os.path.join(
            os.path.dirname(__file__),
            "frontend",
            "index.html"
        )
    )
'''

if old_root in content:
    content = content.replace(old_root, new_root)
else:
    print("⚠️ Existing root endpoint format differs; no replacement made.")

# Add frontend asset mounting at the end of the file
mount_code = '''

# --------------------------------------------------
# React frontend static files
# --------------------------------------------------

FRONTEND_DIR = os.path.join(
    os.path.dirname(__file__),
    "frontend"
)

app.mount(
    "/assets",
    StaticFiles(directory=os.path.join(FRONTEND_DIR, "assets")),
    name="assets"
)
'''

if 'name="assets"' not in content:
    content += mount_code

APP_PATH.write_text(content)

print("✅ FastAPI frontend integration added!")
print("Backend API preserved.")
print("React frontend directory connected.")
```

```python
import os
import zipfile
import shutil

# Project paths
PROJECT_DIR = "/content/cardioseg_frontend"
SRC_DIR = os.path.join(PROJECT_DIR, "src")
BACKUP_DIR = os.path.join(PROJECT_DIR, "backup_before_reference_ui")
EXTRACT_DIR = "/content/cardioseg_reference_extract"

# 1. Find the uploaded ZIP
zip_path = "/content/CardioSeg_AI_Reference_UI_Source.zip"

if not os.path.isfile(zip_path):
    raise FileNotFoundError(
        "UI ZIP not found. Please upload CardioSeg_AI_Reference_UI_Source.zip first."
    )

# 2. Verify the React project
if not os.path.isdir(SRC_DIR):
    raise FileNotFoundError(
        f"React source folder not found: {SRC_DIR}"
    )

# 3. Extract the ZIP
os.makedirs(EXTRACT_DIR, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as archive:
    if archive.testzip() is not None:
        raise ValueError("ZIP archive is damaged.")
    archive.extractall(EXTRACT_DIR)

# 4. Locate the new UI files
found_files = {}

for root, dirs, files_list in os.walk(EXTRACT_DIR):
    for filename in ["App.tsx", "index.css"]:
        if filename in files_list and filename not in found_files:
            found_files[filename] = os.path.join(root, filename)

for filename in ["App.tsx", "index.css"]:
    if filename not in found_files:
        raise FileNotFoundError(
            f"{filename} was not found inside the uploaded ZIP."
        )

# 5. Back up existing files
os.makedirs(BACKUP_DIR, exist_ok=True)

for filename in ["App.tsx", "index.css"]:
    current = os.path.join(SRC_DIR, filename)
    if os.path.isfile(current):
        shutil.copy2(
            current,
            os.path.join(BACKUP_DIR, filename)
        )

# 6. Install the new UI files
for filename in ["App.tsx", "index.css"]:
    destination = os.path.join(SRC_DIR, filename)
    shutil.copy2(found_files[filename], destination)
    print(f"Installed: {filename}")

print("\n✅ Reference UI installed successfully!")
print(f"📁 React project: {PROJECT_DIR}")
print(f"🛡️ Backups saved in: {BACKUP_DIR}")
```

```python

import numpy as np

# Use your existing trained model and validation generator
class_names = {
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

dice_scores = {name: [] for name in class_names.values()}

for images, masks in val_generator:
    predictions = model.predict(images, verbose=0)

    # Convert one-hot masks and predicted probabilities to class labels
    true_labels = np.argmax(masks, axis=-1)
    pred_labels = np.argmax(predictions, axis=-1)

    # Calculate Dice separately for each foreground class
    for class_id, class_name in class_names.items():
        true_mask = (true_labels == class_id)
        pred_mask = (pred_labels == class_id)

        intersection = np.logical_and(true_mask, pred_mask).sum()
        denominator = true_mask.sum() + pred_mask.sum()

        if denominator > 0:
            dice = (2.0 * intersection) / denominator
            dice_scores[class_name].append(dice)

# Display mean Dice across validation batches
for class_name, scores in dice_scores.items():
    if scores:
        print(f"{class_name} Validation Dice: "
              f"{np.mean(scores) * 100:.2f}%")
    else:
        print(f"{class_name}: No valid scores calculated")
```

```python

import os, glob
import numpy as np
import nibabel as nib
import tensorflow as tf
from scipy.ndimage import zoom

MODEL_PATH = "/content/drive/MyDrive/CardioSeg_AI/models/CardioSeg_AI_160_Boundary_best.keras"
TRAIN_DIR = "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/training"

model = tf.keras.models.load_model(MODEL_PATH, compile=False)
H, W = model.input_shape[1:3]

classes = {1: "RV", 2: "Myocardium", 3: "LV"}
scores = {name: [] for name in classes.values()}

for patient_num in range(71, 86):
    patient_dir = os.path.join(TRAIN_DIR, f"patient{patient_num:03d}")
    for mask_path in glob.glob(os.path.join(patient_dir, "*_gt.nii.gz")):
        image_path = mask_path.replace("_gt.nii.gz", ".nii.gz")
        if not os.path.exists(image_path):
            continue

        image = nib.load(image_path).get_fdata()
        mask = nib.load(mask_path).get_fdata().astype(np.uint8)

        for z in range(image.shape[2]):
            x = image[:, :, z]
            y = mask[:, :, z]

            x = (x - x.min()) / (x.max() - x.min() + 1e-8)
            x = zoom(x, (H / x.shape[0], W / x.shape[1]), order=1)
            y = zoom(y, (H / y.shape[0], W / y.shape[1]), order=0)

            pred = np.argmax(model.predict(x[None, ..., None], verbose=0)[0], axis=-1)

            for cid, name in classes.items():
                truth = y == cid
                guessed = pred == cid
                denom = truth.sum() + guessed.sum()
                if denom:
                    scores[name].append(
                        2 * np.logical_and(truth, guessed).sum() / denom
                    )

print("\nVALIDATION RESULTS — MEAN SLICE DICE")
for name, values in scores.items():
    print(f"{name}: {np.mean(values) * 100:.2f}%" if values
          else f"{name}: No scores calculated")
```

```python

import os
import glob
import numpy as np
import nibabel as nib
import matplotlib.pyplot as plt
from scipy.ndimage import zoom
import tensorflow as tf

TEST_DIR = "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/testing"
MODEL_PATH = "/content/drive/MyDrive/CardioSeg_AI/models/CardioSeg_AI_160_Boundary_best.keras"

# Load model only if not already loaded
if "model" not in globals():
    model = tf.keras.models.load_model(MODEL_PATH, compile=False)

patient_id = "patient101"
patient_dir = os.path.join(TEST_DIR, patient_id)

# Find a 3D MRI volume
image_files = [
    f for f in glob.glob(os.path.join(patient_dir, "*.nii*"))
    if "_gt" not in os.path.basename(f)
    and "_4d" not in os.path.basename(f)
]

if not image_files:
    raise FileNotFoundError(f"No 3D MRI volume found in {patient_dir}")

image_path = image_files[0]
volume = nib.load(image_path).get_fdata()

# Select middle slice
slice_idx = volume.shape[2] // 2
img = volume[:, :, slice_idx].astype(np.float32)

# Normalize and resize to model input
H, W = model.input_shape[1:3]
img_norm = (img - img.min()) / (img.max() - img.min() + 1e-8)
img_resized = zoom(img_norm, (H / img.shape[0], W / img.shape[1]), order=1)

# Predict segmentation
prediction = model.predict(img_resized[None, ..., None], verbose=0)[0]
pred_mask = np.argmax(prediction, axis=-1)

# Display MRI and predicted mask overlay
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray", origin="lower")
plt.title(f"{patient_id}: Original MRI")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_resized, cmap="gray", origin="lower")
plt.imshow(
    pred_mask,
    cmap=plt.matplotlib.colors.ListedColormap(
        ["black", "blue", "red", "green"]
    ),
    alpha=0.45,
    origin="lower",
    vmin=0,
    vmax=3
)
plt.title(f"{patient_id}: CardioSeg AI Prediction")
plt.axis("off")

plt.tight_layout()
plt.show()

print("MRI file:", os.path.basename(image_path))
print("Slice:", slice_idx)
print("Predicted classes:", np.unique(pred_mask))
```

```python

import os, glob
import numpy as np
import nibabel as nib
import tensorflow as tf
import matplotlib.pyplot as plt
from scipy.ndimage import zoom
from tqdm.auto import tqdm

TEST_DIR = "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/testing"
MODEL_PATH = "/content/drive/MyDrive/CardioSeg_AI/models/CardioSeg_AI_160_Boundary_best.keras"
OUT_DIR = "/content/drive/MyDrive/CardioSeg_AI/results/official_test_predictions"

os.makedirs(OUT_DIR, exist_ok=True)

# Load the trained model only if needed
if "model" not in globals():
    model = tf.keras.models.load_model(MODEL_PATH, compile=False)

H, W = model.input_shape[1:3]

patients = sorted(
    d for d in os.listdir(TEST_DIR)
    if d.startswith("patient") and os.path.isdir(os.path.join(TEST_DIR, d))
)

print("Test patients:", len(patients))
print("Saving results to:", OUT_DIR)

cmap = plt.matplotlib.colors.ListedColormap(
    ["black", "blue", "red", "green"]
)

completed = 0
failed = []

for patient in tqdm(patients, desc="Processing patients"):
    patient_dir = os.path.join(TEST_DIR, patient)
    patient_out = os.path.join(OUT_DIR, patient)
    os.makedirs(patient_out, exist_ok=True)

    volumes = sorted([
        f for f in glob.glob(os.path.join(patient_dir, "*.nii*"))
        if "_gt" not in os.path.basename(f)
        and "_4d" not in os.path.basename(f)
    ])

    if not volumes:
        failed.append((patient, "No 3D MRI volume found"))
        continue

    try:
        for image_path in volumes:
            nii = nib.load(image_path)
            volume = nii.get_fdata(dtype=np.float32)
            predicted_volume = np.zeros(volume.shape, dtype=np.uint8)

            for z in range(volume.shape[2]):
                img = volume[:, :, z]

                img_min, img_max = np.min(img), np.max(img)
                img = (img - img_min) / (img_max - img_min + 1e-8)

                resized = zoom(
                    img,
                    (H / img.shape[0], W / img.shape[1]),
                    order=1
                )

                probabilities = model.predict(
                    resized[None, ..., None], verbose=0
                )[0]

                pred = np.argmax(probabilities, axis=-1).astype(np.uint8)

                # Resize predicted labels back to original slice dimensions
                pred_original = zoom(
                    pred,
                    (volume.shape[0] / H, volume.shape[1] / W),
                    order=0
                )

                predicted_volume[:, :, z] = pred_original[
                    :volume.shape[0], :volume.shape[1]
                ]

                # Save an overlay PNG for each slice
                plt.figure(figsize=(8, 4))
                plt.imshow(volume[:, :, z], cmap="gray", origin="lower")
                plt.imshow(
                    pred,
                    cmap=cmap, alpha=0.45,
                    vmin=0, vmax=3, origin="lower"
                )
                plt.title(f"{patient} | {os.path.basename(image_path)} | Slice {z}")
                plt.axis("off")
                plt.tight_layout()

                png_name = (
                    os.path.basename(image_path).replace(".nii.gz", "")
                    + f"_slice{z:03d}.png"
                )
                plt.savefig(
                    os.path.join(patient_out, png_name),
                    dpi=100, bbox_inches="tight"
                )
                plt.close()

            # Save predicted mask as NIfTI in original volume geometry
            mask_name = (
                os.path.basename(image_path).replace(".nii.gz", "")
                + "_pred_mask.nii.gz"
            )
            mask_nii = nib.Nifti1Image(
                predicted_volume, nii.affine, nii.header
            )
            nib.save(mask_nii, os.path.join(patient_out, mask_name))

        completed += 1
        print(f"Completed {patient}: {completed}/{len(patients)}")

    except Exception as e:
        failed.append((patient, str(e)))
        print(f"Failed {patient}: {e}")

print("\nINFERENCE SUMMARY")
print("Patients completed:", completed, "/", len(patients))
print("Patients with errors:", len(failed))
print("Output folder:", OUT_DIR)

if failed:
    print("\nErrors:")
    for patient, error in failed:
        print(patient, ":", error)
```

```python

import os
import glob
import numpy as np
import nibabel as nib
import tensorflow as tf
from scipy.ndimage import zoom
from tqdm.auto import tqdm

TEST_DIR = "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/testing"
MODEL_PATH = "/content/drive/MyDrive/CardioSeg_AI/models/CardioSeg_AI_160_Boundary_best.keras"

model = tf.keras.models.load_model(MODEL_PATH, compile=False)
H, W = model.input_shape[1:3]

classes = {1: "RV", 2: "Myocardium", 3: "LV"}
dice_scores = {name: [] for name in classes.values()}
evaluated = 0
skipped = []

mask_paths = sorted(glob.glob(
    os.path.join(TEST_DIR, "**", "*_gt.nii.gz"),
    recursive=True
))

for mask_path in tqdm(mask_paths, desc="Evaluating labelled test frames"):
    image_path = mask_path.replace("_gt.nii.gz", ".nii.gz")

    if not os.path.exists(image_path):
        skipped.append(mask_path)
        continue

    image_nii = nib.load(image_path)
    mask_nii = nib.load(mask_path)

    image_vol = image_nii.get_fdata(dtype=np.float32)
    mask_vol = mask_nii.get_fdata().astype(np.uint8)

    if image_vol.shape != mask_vol.shape:
        skipped.append(mask_path + " (shape mismatch)")
        continue

    # Predict each slice in the labelled frame
    for z in range(image_vol.shape[2]):
        img = image_vol[:, :, z]
        gt = mask_vol[:, :, z]

        img = (img - img.min()) / (img.max() - img.min() + 1e-8)
        img_resized = zoom(
            img, (H / img.shape[0], W / img.shape[1]), order=1
        )

        probs = model.predict(img_resized[None, ..., None], verbose=0)[0]
        pred = np.argmax(probs, axis=-1)

        # Resize ground truth to prediction dimensions using nearest-neighbor
        gt_resized = zoom(
            gt, (H / gt.shape[0], W / gt.shape[1]), order=0
        )

        for class_id, name in classes.items():
            true_mask = gt_resized == class_id
            pred_mask = pred == class_id
            denom = true_mask.sum() + pred_mask.sum()

            # Include empty-empty slices as perfect overlap
            score = (
                1.0 if denom == 0 else
                2.0 * np.logical_and(true_mask, pred_mask).sum() / denom
            )
            dice_scores[name].append(score)

    evaluated += 1

print("\nTEST-SET RESULTS")
print("Reference frames evaluated:", evaluated)
print("Frames skipped:", len(skipped))

for name, values in dice_scores.items():
    if values:
        print(f"{name} mean slice Dice: {np.mean(values) * 100:.2f}%")

means = [
    np.mean(values) for values in dice_scores.values() if values
]
if means:
    print(f"Mean of class-wise Dice: {np.mean(means) * 100:.2f}%")
```

```python

import matplotlib.pyplot as plt
import numpy as np
import nibabel as nib
from scipy.ndimage import zoom

patient_dir = (
    "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/testing/"
    "patient101"
)

image = nib.load(patient_dir + "/patient101_frame01.nii.gz").get_fdata()
mask = nib.load(patient_dir + "/patient101_frame01_gt.nii.gz").get_fdata()

z = 5
img = image[:, :, z].astype(np.float32)
gt = mask[:, :, z].astype(np.uint8)

H, W = model.input_shape[1:3]

# Match the current evaluation preprocessing
img_norm = (img - img.min()) / (img.max() - img.min() + 1e-8)
img_resized = zoom(
    img_norm, (H / img.shape[0], W / img.shape[1]), order=1
)
gt_resized = zoom(
    gt, (H / gt.shape[0], W / gt.shape[1]), order=0
).astype(np.uint8)

pred = np.argmax(
    model.predict(img_resized[None, ..., None], verbose=0)[0],
    axis=-1
)

colors = ["black", "blue", "red", "green"]
cmap = plt.matplotlib.colors.ListedColormap(colors)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(img_resized, cmap="gray", origin="lower")
axes[0].set_title("Original MRI")

axes[1].imshow(gt_resized, cmap=cmap, vmin=0, vmax=3, origin="lower")
axes[1].set_title("Ground Truth")

axes[2].imshow(img_resized, cmap="gray", origin="lower")
axes[2].imshow(
    pred, cmap=cmap, vmin=0, vmax=3, alpha=0.5, origin="lower"
)
axes[2].set_title("CardioSeg AI Prediction")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

import numpy as np

class_names = {
    1: "RV",
    2: "Myocardium",
    3: "LV"
}

print("PATIENT101 — FRAME01 — SLICE 5")
print("-" * 40)

for class_id, name in class_names.items():
    gt_class = (gt_resized == class_id)
    pred_class = (pred == class_id)

    intersection = np.logical_and(gt_class, pred_class).sum()
    denominator = gt_class.sum() + pred_class.sum()

    dice = 1.0 if denominator == 0 else 2 * intersection / denominator

    print(f"{name} Dice: {dice * 100:.2f}%")
```

```python

import os
import glob
import numpy as np
import nibabel as nib
import pandas as pd
from scipy.ndimage import zoom
from tqdm.auto import tqdm

TEST_DIR = "/content/drive/MyDrive/CardioSeg_AI/dataset/ACDC/database/testing"

classes = {1: "RV", 2: "Myocardium", 3: "LV"}
patient_results = []

H, W = model.input_shape[1:3]

mask_paths = sorted(glob.glob(
    os.path.join(TEST_DIR, "**", "*_gt.nii.gz"),
    recursive=True
))

# Collect Dice per patient across its two labelled frames and all slices
for patient in tqdm(sorted(os.listdir(TEST_DIR)), desc="Evaluating patients"):
    patient_dir = os.path.join(TEST_DIR, patient)
    if not os.path.isdir(patient_dir) or not patient.startswith("patient"):
        continue

    patient_masks = sorted(glob.glob(
        os.path.join(patient_dir, "*_gt.nii.gz")
    ))

    class_intersections = {name: 0 for name in classes.values()}
    class_denominators = {name: 0 for name in classes.values()}

    for mask_path in patient_masks:
        image_path = mask_path.replace("_gt.nii.gz", ".nii.gz")
        if not os.path.exists(image_path):
            continue

        image = nib.load(image_path).get_fdata(dtype=np.float32)
        gt_volume = nib.load(mask_path).get_fdata().astype(np.uint8)

        for z in range(image.shape[2]):
            img = image[:, :, z]
            gt = gt_volume[:, :, z]

            img = (img - img.min()) / (img.max() - img.min() + 1e-8)
            img = zoom(img, (H / img.shape[0], W / img.shape[1]), order=1)
            gt = zoom(gt, (H / gt.shape[0], W / gt.shape[1]), order=0)

            pred = np.argmax(
                model.predict(img[None, ..., None], verbose=0)[0],
                axis=-1
            )

            for class_id, name in classes.items():
                true_mask = gt == class_id
                pred_mask = pred == class_id

                class_intersections[name] += np.logical_and(
                    true_mask, pred_mask
                ).sum()
                class_denominators[name] += true_mask.sum() + pred_mask.sum()

    row = {"Patient": patient}

    for name in classes.values():
        denom = class_denominators[name]
        row[f"{name} Dice (%)"] = (
            100 * 2 * class_intersections[name] / denom
            if denom else np.nan
        )

    patient_results.append(row)

df = pd.DataFrame(patient_results)

score_columns = [
    "RV Dice (%)", "Myocardium Dice (%)", "LV Dice (%)"
]
df["Patient Mean Dice (%)"] = df[score_columns].mean(axis=1)

print("\nMEAN DICE BY PATIENT")
display(df.round(2))

print("\nOVERALL MEAN OF PATIENT SCORES")
print(df[score_columns + ["Patient Mean Dice (%)"]].mean().round(2))

output_csv = (
    "/content/drive/MyDrive/CardioSeg_AI/results/"
    "official_test_patient_dice.csv"
)
os.makedirs(os.path.dirname(output_csv), exist_ok=True)
df.to_csv(output_csv, index=False)

print("\nSaved results to:", output_csv)
```
